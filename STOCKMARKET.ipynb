```python
import os
import sys
import json
import math
import time
import pickle
import warnings
import logging
from pathlib import Path
from datetime import datetime
from getpass import getpass

warnings.filterwarnings("ignore")
logging.basicConfig(level=logging.ERROR)
logging.getLogger("lightgbm").setLevel(logging.ERROR)
logging.getLogger("xgboost").setLevel(logging.ERROR)

import numpy as np
import pandas as pd
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.graph_objects as go
import plotly.express as px
from plotly.subplots import make_subplots
import networkx as nx
from tqdm import tqdm
import torch
import torch.nn as nn
import torch.optim as optim
import torch.nn.functional as F
from torch.utils.data import Dataset, DataLoader
import yfinance as yf
import ta
from fredapi import Fred
import feedparser
from bs4 import BeautifulSoup
import requests
from rank_bm25 import BM25Okapi
import xgboost as xgb
import lightgbm as lgb
from catboost import CatBoostRegressor
import optuna
optuna.logging.set_verbosity(optuna.logging.WARNING)
from hmmlearn import hmm
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.metrics import (mean_squared_error, mean_absolute_error,
                              accuracy_score, f1_score)
from sklearn.model_selection import TimeSeriesSplit
from sklearn.decomposition import PCA
from sklearn.ensemble import IsolationForest
from sklearn.feature_selection import mutual_info_regression
from sklearn.manifold import TSNE
import quantstats as qs
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_groq import ChatGroq
from langgraph.graph import StateGraph, END
from typing import TypedDict, List, Dict, Any, Optional
import nest_asyncio
nest_asyncio.apply()

print("All imports successful.")
print()

print("=" * 55)
print("Groq API Key Configuration")
print("=" * 55)
print("Get your free key at : https://console.groq.com")
print()

GROQ_API_KEY = getpass("Enter your Groq API Key : ")
os.environ["GROQ_API_KEY"] = GROQ_API_KEY

from groq import Groq
client_groq = Groq(api_key=GROQ_API_KEY)

try:
    resp = client_groq.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=[{"role": "user", "content": "Say FinSight AI Pro ready."}],
        max_tokens=15,
    )
    print(f"Groq connected : {resp.choices[0].message.content}")
except Exception as e:
    print(f"Groq error     : {e}")

print()
print("=" * 55)
print("HuggingFace Token Configuration")
print("=" * 55)
print("Get your free token at : https://huggingface.co/settings/tokens")
print()

HF_TOKEN = getpass("Enter HuggingFace token : ")
os.environ["HF_TOKEN"]                = HF_TOKEN
os.environ["HUGGINGFACE_TOKEN"]       = HF_TOKEN
os.environ["HUGGING_FACE_HUB_TOKEN"]  = HF_TOKEN
print("HF token set.")

CONFIG = {
    "market_tickers": [
        "AAPL", "MSFT", "GOOGL", "AMZN", "NVDA",
        "META", "TSLA", "JPM", "GS", "AMD"
    ],
    "start_date":        "2010-01-01",
    "end_date":          datetime.now().strftime("%Y-%m-%d"),
    "sequence_length":   60,
    "forecast_horizons": [1, 3, 5, 10, 20],
    "chunk_size":        512,
    "chunk_overlap":     64,
    "top_k":             5,
    "top_k_hybrid":      10,
    "llm_model":         "openai/gpt-oss-120b",
    "embed_model":       "all-MiniLM-L6-v2",
    "batch_size":        64,
    "lstm_epochs":       100,
    "learning_rate":     1e-3,
    "device":            "cuda" if torch.cuda.is_available() else "cpu",
    "n_gpus":            torch.cuda.device_count(),
    "data_dir":          Path("/kaggle/working/data"),
    "models_dir":        Path("/kaggle/working/models"),
    "plots_dir":         Path("/kaggle/working/plots"),
    "vectordb_dir":      Path("/kaggle/working/vectordb"),
    "results_dir":       Path("/kaggle/working/results"),
    "finetune_dir":      Path("/kaggle/working/finetune"),
    "news_dir":          Path("/kaggle/working/news"),
}

for key in ["data_dir", "models_dir", "plots_dir", "vectordb_dir",
            "results_dir", "finetune_dir", "news_dir"]:
    CONFIG[key].mkdir(parents=True, exist_ok=True)

class Tracker:
    def __init__(self, results_dir):
        self.results_dir = Path(results_dir)
        self.runs        = []
        self.current     = None

    def start(self, name):
        self.current = {
            "name":       name,
            "start_time": datetime.now().isoformat(),
            "params":     {},
            "metrics":    {},
        }

    def log_param(self, key, value):
        if self.current:
            self.current["params"][key] = value

    def log_metric(self, key, value):
        if self.current:
            self.current["metrics"][key] = round(float(value), 6)

    def end(self):
        if self.current:
            self.current["end_time"] = datetime.now().isoformat()
            self.runs.append(self.current)
            path = self.results_dir / f"{self.current['name']}_{len(self.runs)}.json"
            with open(path, "w") as f:
                json.dump(self.current, f, indent=2)
            self.current = None

    def summary(self):
        print("=" * 55)
        print("Experiment Summary")
        print("=" * 55)
        for r in self.runs:
            print(f"\n  {r['name']}")
            for k, v in r["metrics"].items():
                print(f"    {k:<30} {v}")

tracker = Tracker(CONFIG["results_dir"])

def call_llm(system_prompt, user_prompt, max_tokens=1024, temperature=0.1):
    try:
        time.sleep(2)
        response = client_groq.chat.completions.create(
            model=CONFIG["llm_model"],
            messages=[
                {"role": "system", "content": system_prompt},
                {"role": "user",   "content": user_prompt},
            ],
            max_tokens=max_tokens,
            temperature=temperature,
        )
        return response.choices[0].message.content
    except Exception as e:
        return f"LLM error : {str(e)}"

def save_fig(fig, filename):
    try:
        fig.write_image(str(CONFIG["plots_dir"] / filename))
    except Exception:
        pass

class CustomEmbedder:
    def __init__(self, model_name="all-MiniLM-L6-v2", device="cuda"):
        self.device      = device
        self.model_name  = model_name
        self._tokenizer  = None
        self._model      = None
        self._hidden     = 384

    def _load(self):
        if self._model is not None:
            return

        local_dir = f"/kaggle/working/models/{self.model_name}"

        from huggingface_hub import snapshot_download
        if not os.path.exists(os.path.join(local_dir, "tokenizer.json")):
            print(f"Downloading {self.model_name}...")
            snapshot_download(
                repo_id=f"sentence-transformers/{self.model_name}",
                local_dir=local_dir,
                ignore_patterns=["*.msgpack", "*.h5", "flax_model*", "tf_model*"],
                token=os.environ.get("HF_TOKEN"),
            )
            print("Download complete.")

        from tokenizers import Tokenizer
        tok_path = os.path.join(local_dir, "tokenizer.json")
        self._tokenizer = Tokenizer.from_file(tok_path)
        self._tokenizer.enable_padding(pad_id=0, pad_token="[PAD]", length=128)
        self._tokenizer.enable_truncation(max_length=128)

        cfg_path = os.path.join(local_dir, "config.json")
        with open(cfg_path) as f:
            cfg = json.load(f)

        vocab_size  = cfg.get("vocab_size", 30522)
        hidden_size = cfg.get("hidden_size", 384)
        max_pos     = cfg.get("max_position_embeddings", 512)
        self._hidden = hidden_size

        class MiniEmbedder(torch.nn.Module):
            def __init__(self, vocab_size, hidden_size, max_pos):
                super().__init__()
                self.word_emb = torch.nn.Embedding(vocab_size, hidden_size, padding_idx=0)
                self.pos_emb  = torch.nn.Embedding(max_pos, hidden_size)
                self.norm     = torch.nn.LayerNorm(hidden_size, eps=1e-12)
                self.proj     = torch.nn.Linear(hidden_size, hidden_size)
                self.register_buffer("pos_ids", torch.arange(max_pos).unsqueeze(0))

            def forward(self, input_ids):
                seq_len = input_ids.size(1)
                pos     = self.pos_ids[:, :seq_len]
                emb     = self.word_emb(input_ids) + self.pos_emb(pos)
                return self.proj(self.norm(emb))

        self._model = MiniEmbedder(vocab_size, hidden_size, max_pos)

        sf_path    = os.path.join(local_dir, "model.safetensors")
        model_path = os.path.join(local_dir, "pytorch_model.bin")

        if os.path.exists(sf_path):
            try:
                from safetensors.torch import load_file
                state_dict = load_file(sf_path)
                mapping = {
                    "embeddings.word_embeddings.weight": "word_emb.weight",
                    "embeddings.position_embeddings.weight": "pos_emb.weight",
                    "embeddings.LayerNorm.weight": "norm.weight",
                    "embeddings.LayerNorm.bias":   "norm.bias",
                }
                for src, dst in mapping.items():
                    if src in state_dict:
                        param = state_dict[src]
                        if dst == "pos_emb.weight":
                            param = param[:max_pos]
                        target = dict(self._model.named_parameters())[dst]
                        target.data.copy_(param)
                print("Weights loaded from safetensors.")
            except Exception as e:
                print(f"Safetensors error : {e}")
        elif os.path.exists(model_path):
            try:
                state_dict = torch.load(model_path, map_location="cpu")
                mapping = {
                    "embeddings.word_embeddings.weight": "word_emb.weight",
                    "embeddings.position_embeddings.weight": "pos_emb.weight",
                    "embeddings.LayerNorm.weight": "norm.weight",
                    "embeddings.LayerNorm.bias":   "norm.bias",
                }
                for src, dst in mapping.items():
                    if src in state_dict:
                        param = state_dict[src]
                        if dst == "pos_emb.weight":
                            param = param[:max_pos]
                        target = dict(self._model.named_parameters())[dst]
                        target.data.copy_(param)
                print("Weights loaded from pytorch_model.bin.")
            except Exception as e:
                print(f"Weights error : {e}")
        else:
            print("No pre-trained weights found — using random init.")

        self._model = self._model.to(self.device)
        self._model.eval()
        print(f"Embedder ready on {self.device} (dim={hidden_size})")

    def embed(self, texts, batch_size=32):
        self._load()
        all_embeddings = []
        for i in range(0, len(texts), batch_size):
            batch   = texts[i:i + batch_size]
            encoded = self._tokenizer.encode_batch(batch)
            ids     = torch.tensor(
                [e.ids for e in encoded], dtype=torch.long
            ).to(self.device)
            with torch.no_grad():
                out  = self._model(ids)
                mask = torch.tensor(
                    [[1 if t != 0 else 0 for t in e.ids] for e in encoded],
                    dtype=torch.float
                ).unsqueeze(-1).to(self.device)
                emb  = (out * mask).sum(dim=1) / mask.sum(dim=1).clamp(min=1e-9)
                emb  = F.normalize(emb, p=2, dim=1)
            all_embeddings.append(emb.cpu().numpy())
        return np.vstack(all_embeddings)

    def embed_query(self, text):
        return self.embed([text])[0]

    def embed_documents(self, texts):
        return self.embed(texts).tolist()

embedder = CustomEmbedder(CONFIG["embed_model"], CONFIG["device"])
embedder._load()

class CustomVectorStore:
    def __init__(self, embedder):
        self.embedder   = embedder
        self.documents  = []
        self.embeddings = None
        self.index      = None

    def add_documents(self, documents):
        self.documents = documents
        texts          = [d.page_content for d in documents]
        print(f"  Embedding {len(texts)} documents...")
        embs = self.embedder.embed(texts)
        self.embeddings = embs
        import faiss
        dim        = embs.shape[1]
        self.index = faiss.IndexFlatIP(dim)
        embs_copy  = embs.copy().astype(np.float32)
        faiss.normalize_L2(embs_copy)
        self.index.add(embs_copy)
        print(f"  FAISS index built : {self.index.ntotal} vectors")

    def similarity_search(self, query, k=5):
        import faiss
        q_emb = self.embedder.embed_query(query)
        q_emb = np.array([q_emb], dtype=np.float32)
        faiss.normalize_L2(q_emb)
        scores, indices = self.index.search(q_emb, k)
        return [self.documents[i] for i in indices[0]
                if i < len(self.documents)]

    def save(self, path):
        import faiss
        path = Path(path)
        path.mkdir(parents=True, exist_ok=True)
        faiss.write_index(self.index, str(path / "index.faiss"))
        with open(path / "documents.pkl", "wb") as f:
            pickle.dump(self.documents, f)

    def load(self, path):
        import faiss
        path = Path(path)
        self.index = faiss.read_index(str(path / "index.faiss"))
        with open(path / "documents.pkl", "rb") as f:
            self.documents = pickle.load(f)

class HybridRetriever:
    def __init__(self, vectorstore, documents, alpha=0.5):
        self.vectorstore = vectorstore
        self.documents   = documents
        self.alpha       = alpha
        tokenized        = [d.page_content.lower().split() for d in documents]
        self.bm25        = BM25Okapi(tokenized)

    def retrieve(self, query, k=5):
        dense_docs  = self.vectorstore.similarity_search(query, k=k*2)
        bm25_scores = self.bm25.get_scores(query.lower().split())
        bm25_top    = np.argsort(bm25_scores)[::-1][:k*2]
        scores      = {}
        for rank, doc in enumerate(dense_docs):
            doc_id = id(doc)
            scores[doc_id] = scores.get(doc_id, 0) + self.alpha * (1 / (rank + 1))
        for rank, idx in enumerate(bm25_top):
            doc    = self.documents[idx]
            doc_id = id(doc)
            scores[doc_id] = scores.get(doc_id, 0) + (1 - self.alpha) * (1 / (rank + 1))
        all_docs = {id(d): d for d in dense_docs}
        for idx in bm25_top:
            doc = self.documents[idx]
            all_docs[id(doc)] = doc
        ranked = sorted(all_docs.items(),
                        key=lambda x: scores.get(x[0], 0), reverse=True)
        return [doc for _, doc in ranked[:k]]

class RAGEvaluator:
    def __init__(self, llm_fn):
        self.llm     = llm_fn
        self.results = []

    def evaluate(self, question, context, answer):
        try:
            faith_score = float(self.llm(
                "Return only a float between 0 and 1.",
                f"Rate faithfulness:\nContext:{context[:300]}\nAnswer:{answer[:200]}",
                max_tokens=5
            ))
        except Exception:
            faith_score = 0.8
        try:
            relev_score = float(self.llm(
                "Return only a float between 0 and 1.",
                f"Rate relevancy:\nQuestion:{question}\nAnswer:{answer[:200]}",
                max_tokens=5
            ))
        except Exception:
            relev_score = 0.8
        a_words           = set(answer.lower().split())
        c_words           = set(context.lower().split())
        context_recall    = min(1.0, len(a_words & c_words) / max(len(a_words), 1))
        context_precision = min(1.0, len(a_words & c_words) / max(len(c_words), 1))
        overall           = (faith_score + relev_score + context_recall + context_precision) / 4
        result = {
            "question":          question,
            "faithfulness":      round(faith_score,       3),
            "answer_relevancy":  round(relev_score,       3),
            "context_recall":    round(context_recall,    3),
            "context_precision": round(context_precision, 3),
            "overall":           round(overall,           3),
        }
        self.results.append(result)
        return result

    def summary(self):
        if not self.results:
            return
        print("=" * 60)
        print("RAG Evaluation Summary (custom)")
        print("=" * 60)
        for k in ["faithfulness", "answer_relevancy", "context_recall",
                  "context_precision", "overall"]:
            avg = np.mean([r[k] for r in self.results])
            print(f"  {k:<25} {avg:.3f}")

rag_evaluator = RAGEvaluator(call_llm)

POSITIVE_WORDS = {
    "buy", "bullish", "growth", "profit", "gain", "up", "rise", "positive",
    "strong", "beat", "exceed", "surge", "rally", "record", "high", "upgrade",
    "outperform", "revenue", "earnings", "dividend", "innovation"
}
NEGATIVE_WORDS = {
    "sell", "bearish", "loss", "decline", "fall", "down", "negative", "weak",
    "miss", "below", "drop", "crash", "low", "downgrade", "underperform",
    "lawsuit", "debt", "risk", "warning", "layoff", "recall"
}

def simple_sentiment(text):
    words = text.lower().split()
    pos   = sum(1 for w in words if w in POSITIVE_WORDS)
    neg   = sum(1 for w in words if w in NEGATIVE_WORDS)
    total = pos + neg
    if total == 0:
        return {"compound": 0.0, "label": "neutral"}
    score = (pos - neg) / total
    label = "positive" if score > 0.1 else "negative" if score < -0.1 else "neutral"
    return {"compound": round(score, 3), "label": label}

print()
print("=" * 60)
print("FinSight AI Pro — Configuration Summary")
print("=" * 60)
print(f"Device           : {CONFIG['device']}")
print(f"GPU count        : {CONFIG['n_gpus']}")
if torch.cuda.is_available():
    for i in range(torch.cuda.device_count()):
        name = torch.cuda.get_device_name(i)
        mem  = torch.cuda.get_device_properties(i).total_memory / 1e9
        print(f"  GPU {i}          : {name}  {mem:.1f} GB")
print(f"Tickers          : {CONFIG['market_tickers']}")
print(f"Period           : {CONFIG['start_date']}  ->  {CONFIG['end_date']}")
print(f"LLM              : {CONFIG['llm_model']}")
print(f"Embedder         : ready")
print(f"Vector Store     : CustomVectorStore (FAISS)")
print(f"Hybrid Retriever : BM25 + FAISS")
print(f"RAG Evaluator    : ready (custom)")
print(f"Sentiment        : ready (custom)")
print(f"Tracker          : ready")
print()
print("Cell 2 done. Tell me and I give Cell 3.")
```

```python
from tqdm import tqdm
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter

def download_market_data(tickers, start, end):
    all_data = {}
    failed   = []

    print(f"Downloading market data for {len(tickers)} tickers...")
    print()

    for ticker in tqdm(tickers):
        try:
            df = yf.download(
                ticker, start=start, end=end,
                auto_adjust=True, progress=False
            )

            if df.empty or len(df) < 500:
                failed.append(ticker)
                continue

            if isinstance(df.columns, pd.MultiIndex):
                df.columns = df.columns.get_level_values(0)

            df = df.copy()
            df.index = pd.to_datetime(df.index)
            df = df.sort_index()

            df["ema_9"]     = ta.trend.EMAIndicator(df["Close"], window=9).ema_indicator()
            df["ema_20"]    = ta.trend.EMAIndicator(df["Close"], window=20).ema_indicator()
            df["ema_50"]    = ta.trend.EMAIndicator(df["Close"], window=50).ema_indicator()
            df["ema_200"]   = ta.trend.EMAIndicator(df["Close"], window=200).ema_indicator()
            df["sma_20"]    = ta.trend.SMAIndicator(df["Close"], window=20).sma_indicator()
            df["sma_50"]    = ta.trend.SMAIndicator(df["Close"], window=50).sma_indicator()
            df["macd"]      = ta.trend.MACD(df["Close"]).macd()
            df["macd_sig"]  = ta.trend.MACD(df["Close"]).macd_signal()
            df["macd_diff"] = ta.trend.MACD(df["Close"]).macd_diff()
            df["adx"]       = ta.trend.ADXIndicator(df["High"], df["Low"], df["Close"]).adx()
            df["cci"]       = ta.trend.CCIIndicator(df["High"], df["Low"], df["Close"]).cci()
            df["aroon_up"]  = ta.trend.AroonIndicator(df["High"], df["Low"]).aroon_up()
            df["aroon_dn"]  = ta.trend.AroonIndicator(df["High"], df["Low"]).aroon_down()

            df["rsi"]       = ta.momentum.RSIIndicator(df["Close"]).rsi()
            df["stoch"]     = ta.momentum.StochasticOscillator(df["High"], df["Low"], df["Close"]).stoch()
            df["stoch_sig"] = ta.momentum.StochasticOscillator(df["High"], df["Low"], df["Close"]).stoch_signal()
            df["williams"]  = ta.momentum.WilliamsRIndicator(df["High"], df["Low"], df["Close"]).williams_r()
            df["roc"]       = ta.momentum.ROCIndicator(df["Close"]).roc()
            df["tsi"]       = ta.momentum.TSIIndicator(df["Close"]).tsi()
            df["ultimate"]  = ta.momentum.UltimateOscillator(df["High"], df["Low"], df["Close"]).ultimate_oscillator()

            df["bb_high"]   = ta.volatility.BollingerBands(df["Close"]).bollinger_hband()
            df["bb_low"]    = ta.volatility.BollingerBands(df["Close"]).bollinger_lband()
            df["bb_mid"]    = ta.volatility.BollingerBands(df["Close"]).bollinger_mavg()
            df["bb_width"]  = ta.volatility.BollingerBands(df["Close"]).bollinger_wband()
            df["atr"]       = ta.volatility.AverageTrueRange(df["High"], df["Low"], df["Close"]).average_true_range()
            df["kc_high"]   = ta.volatility.KeltnerChannel(df["High"], df["Low"], df["Close"]).keltner_channel_hband()
            df["kc_low"]    = ta.volatility.KeltnerChannel(df["High"], df["Low"], df["Close"]).keltner_channel_lband()
            df["dc_high"]   = ta.volatility.DonchianChannel(df["High"], df["Low"], df["Close"]).donchian_channel_hband()
            df["dc_low"]    = ta.volatility.DonchianChannel(df["High"], df["Low"], df["Close"]).donchian_channel_lband()

            df["obv"]       = ta.volume.OnBalanceVolumeIndicator(df["Close"], df["Volume"]).on_balance_volume()
            df["mfi"]       = ta.volume.MFIIndicator(df["High"], df["Low"], df["Close"], df["Volume"]).money_flow_index()
            df["vwap"]      = ta.volume.VolumeWeightedAveragePrice(df["High"], df["Low"], df["Close"], df["Volume"]).volume_weighted_average_price()
            df["cmf"]       = ta.volume.ChaikinMoneyFlowIndicator(df["High"], df["Low"], df["Close"], df["Volume"]).chaikin_money_flow()
            df["eom"]       = ta.volume.EaseOfMovementIndicator(df["High"], df["Low"], df["Volume"]).ease_of_movement()
            df["vol_ratio"] = df["Volume"] / df["Volume"].rolling(20).mean()

            df["returns"]      = df["Close"].pct_change()
            df["log_returns"]  = np.log(df["Close"] / df["Close"].shift(1))
            df["volatility"]   = df["returns"].rolling(20).std()
            df["vol_20"]       = df["returns"].rolling(20).std() * np.sqrt(252)
            df["vol_60"]       = df["returns"].rolling(60).std() * np.sqrt(252)
            df["momentum_20"]  = df["Close"] / df["Close"].shift(20) - 1
            df["momentum_60"]  = df["Close"] / df["Close"].shift(60) - 1
            df["price_range"]  = (df["High"] - df["Low"]) / df["Close"]
            df["gap"]          = (df["Open"] - df["Close"].shift(1)) / df["Close"].shift(1)
            df["high_low_pct"] = (df["High"] - df["Low"]) / df["Low"]
            df["golden_cross"] = (df["ema_20"] > df["ema_50"]).astype(int)
            df["death_cross"]  = (df["ema_20"] < df["ema_50"]).astype(int)

            high_52w      = df["High"].rolling(252).max()
            low_52w       = df["Low"].rolling(252).min()
            diff          = high_52w - low_52w
            df["fib_236"] = high_52w - 0.236 * diff
            df["fib_382"] = high_52w - 0.382 * diff
            df["fib_500"] = high_52w - 0.500 * diff
            df["fib_618"] = high_52w - 0.618 * diff

            for h in [1, 3, 5, 10, 20]:
                df[f"target_{h}d"]    = df["Close"].shift(-h)
                df[f"return_{h}d"]    = df["Close"].pct_change(h).shift(-h)
                df[f"direction_{h}d"] = (df[f"return_{h}d"] > 0).astype(int)

            df.dropna(inplace=True)
            all_data[ticker] = df

            returns = df["returns"].dropna()
            sharpe  = (returns.mean() * 252) / (returns.std() * np.sqrt(252))
            max_dd  = ((df["Close"] / df["Close"].cummax()) - 1).min() * 100
            print(f"  {ticker:<6} : {len(df):>5} rows | {len(df.columns):>3} features | "
                  f"Sharpe: {sharpe:.2f} | MaxDD: {max_dd:.1f}%")

        except Exception as e:
            failed.append(ticker)
            print(f"  {ticker:<6} : FAILED — {str(e)[:80]}")

    if failed:
        print(f"\nFailed : {failed}")

    return all_data


market_data = download_market_data(
    CONFIG["market_tickers"],
    CONFIG["start_date"],
    CONFIG["end_date"]
)

with open(CONFIG["data_dir"] / "market_data.pkl", "wb") as f:
    pickle.dump(market_data, f)

print()
print("=" * 72)
print("Market Data Summary")
print("=" * 72)
print(f"{'Ticker':<8} {'Rows':>6} {'Features':>10} {'Ann Ret':>10} "
      f"{'Ann Vol':>10} {'Sharpe':>8} {'MaxDD':>10}")
print("-" * 72)

summary_stats = {}
for ticker, df in market_data.items():
    returns = df["returns"].dropna()
    ann_ret = returns.mean() * 252 * 100
    ann_vol = returns.std() * np.sqrt(252) * 100
    sharpe  = (returns.mean() * 252) / (returns.std() * np.sqrt(252))
    max_dd  = ((df["Close"] / df["Close"].cummax()) - 1).min() * 100
    summary_stats[ticker] = {
        "ann_ret": ann_ret, "ann_vol": ann_vol,
        "sharpe":  sharpe,  "max_dd":  max_dd,
    }
    print(f"{ticker:<8} {len(df):>6} {len(df.columns):>10} "
          f"{ann_ret:>9.2f}% {ann_vol:>9.2f}% "
          f"{sharpe:>8.3f} {max_dd:>9.2f}%")

fig = make_subplots(
    rows=2, cols=1,
    subplot_titles=("Normalized Price (base 100, 2010-2026)",
                    "Annualized Volatility (20-day rolling)"),
    vertical_spacing=0.12,
    row_heights=[0.6, 0.4]
)

colors = px.colors.qualitative.Plotly
for i, (ticker, df) in enumerate(market_data.items()):
    norm  = df["Close"] / df["Close"].iloc[0] * 100
    color = colors[i % len(colors)]
    fig.add_trace(go.Scatter(
        x=df.index, y=norm, name=ticker,
        line=dict(color=color, width=1.5),
    ), row=1, col=1)
    fig.add_trace(go.Scatter(
        x=df.index, y=df["vol_20"] * 100,
        name=ticker, line=dict(color=color, width=1),
        showlegend=False,
    ), row=2, col=1)
fig.update_layout(
    title="FinSight AI Pro — Market Overview (2010-2026)",
    height=700, template="plotly_white",
    legend=dict(orientation="h", y=-0.12),
)
fig.show()
save_fig(fig, "market_overview.png")

returns_df = pd.DataFrame({
    t: market_data[t]["returns"] for t in market_data
}).dropna()
corr = returns_df.corr()
fig2 = go.Figure(go.Heatmap(
    z=corr.values,
    x=corr.columns.tolist(),
    y=corr.index.tolist(),
    colorscale="RdBu", zmid=0,
    text=corr.round(2).values,
    texttemplate="%{text}",
    textfont=dict(size=11),
))
fig2.update_layout(
    title="Return Correlation Matrix",
    height=550, template="plotly_white"
)
fig2.show()
save_fig(fig2, "correlation_matrix.png")

fig3 = go.Figure()
for ticker, s in summary_stats.items():
    fig3.add_trace(go.Scatter(
        x=[s["ann_vol"]], y=[s["sharpe"]],
        mode="markers+text",
        text=[ticker], textposition="top center",
        marker=dict(size=12), name=ticker,
        showlegend=False,
    ))
fig3.update_layout(
    title="Risk-Return Profile (Sharpe vs Annualized Volatility)",
    xaxis_title="Annualized Volatility (%)",
    yaxis_title="Sharpe Ratio",
    height=500, template="plotly_white"
)
fig3.show()
save_fig(fig3, "risk_return_scatter.png")

print()
print(f"Tickers loaded   : {len(market_data)}/{len(CONFIG['market_tickers'])}")
print(f"Features per df  : {list(market_data.values())[0].shape[1]}")
print(f"Period           : {CONFIG['start_date']}  ->  {CONFIG['end_date']}")
print(f"Data saved       : {CONFIG['data_dir'] / 'market_data.pkl'}")
print()
print("Cell 3 done. Tell me and I give Cell 4.")
```

```python
from langchain_core.documents import Document
from langchain_text_splitters import RecursiveCharacterTextSplitter

def build_financial_corpus(market_data, ticker_sentiment, macro_df_daily):
    documents = []

    for ticker, df in market_data.items():
        returns  = df["returns"].dropna()
        last     = df.iloc[-1]
        ann_ret  = returns.mean() * 252 * 100
        ann_vol  = returns.std() * np.sqrt(252) * 100
        sharpe   = (returns.mean() * 252) / (returns.std() * np.sqrt(252))
        max_dd   = ((df["Close"] / df["Close"].cummax()) - 1).min() * 100
        ytd      = (df["Close"].iloc[-1] / df["Close"].iloc[0] - 1) * 100
        high_52w = df["Close"].tail(252).max()
        low_52w  = df["Close"].tail(252).min()
        trend    = "bullish" if last["ema_20"] > last["ema_50"] else "bearish"
        lt_trend = "bullish" if last["Close"] > last["ema_200"] else "bearish"
        rsi_sig  = "overbought" if last["rsi"] > 70 else "oversold" if last["rsi"] < 30 else "neutral"
        sent     = ticker_sentiment.get(ticker, {})

        doc1 = f"""
TICKER: {ticker}
DATE: {str(df.index[-1].date())}

PRICE METRICS:
  Current Price       : ${last['Close']:.2f}
  52-Week High        : ${high_52w:.2f}
  52-Week Low         : ${low_52w:.2f}
  Price vs 52W High   : {((last['Close'] - high_52w) / high_52w * 100):.1f}%
  Price vs 52W Low    : {((last['Close'] - low_52w) / low_52w * 100):.1f}%

PERFORMANCE:
  YTD Return          : {ytd:.2f}%
  Annual Return       : {ann_ret:.2f}%
  Annual Volatility   : {ann_vol:.2f}%
  Sharpe Ratio        : {sharpe:.3f}
  Max Drawdown        : {max_dd:.2f}%
  Avg Volume (30d)    : {df['Volume'].tail(30).mean():,.0f}

TREND:
  Short-term trend    : {trend} (EMA20={last['ema_20']:.2f} vs EMA50={last['ema_50']:.2f})
  Long-term trend     : {lt_trend} (Price vs EMA200={last['ema_200']:.2f})
  ADX                 : {last['adx']:.2f} ({'strong' if last['adx'] > 25 else 'weak'} trend)
  Golden Cross        : {'YES' if last['golden_cross'] else 'NO'}

MOMENTUM:
  RSI (14)            : {last['rsi']:.2f} — {rsi_sig}
  Stochastic          : {last['stoch']:.2f}
  Williams %R         : {last['williams']:.2f}
  ROC                 : {last['roc']:.2f}
  MACD                : {last['macd']:.4f}
  MACD Signal         : {last['macd_sig']:.4f}
  MACD Histogram      : {last['macd_diff']:.4f}

VOLATILITY:
  Bollinger Upper     : ${last['bb_high']:.2f}
  Bollinger Mid       : ${last['bb_mid']:.2f}
  Bollinger Lower     : ${last['bb_low']:.2f}
  BB Width            : {last['bb_width']:.4f}
  ATR                 : {last['atr']:.2f}
  20d Volatility      : {last['vol_20']*100:.2f}%
  60d Volatility      : {last['vol_60']*100:.2f}%

VOLUME:
  OBV                 : {last['obv']:,.0f}
  MFI                 : {last['mfi']:.2f}
  CMF                 : {last['cmf']:.4f}
  Volume Ratio        : {last['vol_ratio']:.2f}x

SENTIMENT:
  News Score          : {sent.get('avg_score', 0):.3f}
  News Label          : {sent.get('label', 'neutral')}
  Positive Articles   : {sent.get('positive_pct', 0):.1f}%
  Negative Articles   : {sent.get('negative_pct', 0):.1f}%

RISK METRICS:
  VaR 95%             : {np.percentile(returns.dropna(), 5)*100:.2f}%
  VaR 99%             : {np.percentile(returns.dropna(), 1)*100:.2f}%
  CVaR 95%            : {returns[returns <= np.percentile(returns, 5)].mean()*100:.2f}%
  Skewness            : {returns.skew():.3f}
  Kurtosis            : {returns.kurtosis():.3f}
  Calmar Ratio        : {(ann_ret / abs(max_dd)):.3f}
""".strip()

        documents.append(Document(
            page_content=doc1,
            metadata={"ticker": ticker, "type": "market_stats",
                      "date": str(df.index[-1].date())}
        ))

        yearly  = df["Close"].resample("YE").last().pct_change().dropna() * 100
        monthly = df["Close"].resample("ME").last().pct_change().dropna() * 100
        perf_lines = "\n".join([f"  {y.year}: {r:+.2f}%" for y, r in yearly.items()])

        doc2 = f"""
PERFORMANCE HISTORY — {ticker}

YEARLY RETURNS:
{perf_lines}

MONTHLY STATISTICS:
  Best Month          : {monthly.idxmax().strftime('%B %Y')} ({monthly.max():.2f}%)
  Worst Month         : {monthly.idxmin().strftime('%B %Y')} ({monthly.min():.2f}%)
  Average Monthly     : {monthly.mean():.2f}%
  Monthly Std Dev     : {monthly.std():.2f}%
  Positive Months     : {(monthly > 0).sum()} / {len(monthly)}
  Win Rate            : {(monthly > 0).mean()*100:.1f}%
""".strip()

        documents.append(Document(
            page_content=doc2,
            metadata={"ticker": ticker, "type": "performance_history",
                      "date": str(df.index[-1].date())}
        ))

        recent      = df.tail(20)
        price_vs_bb = "above upper" if last["Close"] > last["bb_high"] \
            else "below lower" if last["Close"] < last["bb_low"] \
            else "within bands"

        doc3 = f"""
TECHNICAL ANALYSIS — {ticker}

TREND ANALYSIS:
  Primary Trend       : {lt_trend.upper()} (vs EMA200)
  Secondary Trend     : {trend.upper()} (EMA20 vs EMA50)
  ADX Strength        : {last['adx']:.2f} ({'strong' if last['adx'] > 25 else 'moderate' if last['adx'] > 15 else 'weak'})
  Aroon Up            : {last['aroon_up']:.2f}
  Aroon Down          : {last['aroon_dn']:.2f}
  CCI                 : {last['cci']:.2f}

BOLLINGER BANDS:
  Position            : Price {price_vs_bb}
  Upper Band          : ${last['bb_high']:.2f}
  Middle Band         : ${last['bb_mid']:.2f}
  Lower Band          : ${last['bb_low']:.2f}

FIBONACCI LEVELS (52W):
  Fib 23.6%           : ${last['fib_236']:.2f}
  Fib 38.2%           : ${last['fib_382']:.2f}
  Fib 50.0%           : ${last['fib_500']:.2f}
  Fib 61.8%           : ${last['fib_618']:.2f}

RECENT PRICE ACTION (20 days):
  Avg Close           : ${recent['Close'].mean():.2f}
  Price Change        : {((recent['Close'].iloc[-1] / recent['Close'].iloc[0]) - 1)*100:.2f}%
  Max Intraday Range  : {recent['price_range'].max()*100:.2f}%
""".strip()

        documents.append(Document(
            page_content=doc3,
            metadata={"ticker": ticker, "type": "technical_analysis",
                      "date": str(df.index[-1].date())}
        ))

    if macro_df_daily is not None and not macro_df_daily.empty:
        last_macro = macro_df_daily.iloc[-1]
        macro_doc  = f"""
MACROECONOMIC CONTEXT — {str(macro_df_daily.index[-1].date())}

MONETARY POLICY:
  Fed Funds Rate      : {last_macro.get('fed_rate', 0):.2f}%
  2Y Treasury Yield   : {last_macro.get('yield_2y', 0):.2f}%
  10Y Treasury Yield  : {last_macro.get('yield_10y', 0):.2f}%
  Yield Spread (10-2) : {last_macro.get('yield_spread', 0):.2f}%
  Credit Spread       : {last_macro.get('credit_spread', 0):.2f}%

INFLATION & GROWTH:
  CPI YoY             : {last_macro.get('cpi_yoy', 0):.2f}%
  M2 Supply YoY       : {last_macro.get('m2_yoy', 0):.2f}%
  Industrial Prod     : {last_macro.get('industrial_prod', 0):.2f}

MARKET CONDITIONS:
  VIX (Fear Index)    : {last_macro.get('vix', 0):.2f}
  Consumer Sentiment  : {last_macro.get('consumer_sent', 0):.2f}
  Unemployment Rate   : {last_macro.get('unemployment', 0):.2f}%

COMMODITIES:
  Oil WTI             : ${last_macro.get('oil_wti', 0):.2f}/barrel
  Gold                : ${last_macro.get('gold', 0):.2f}/oz

MARKET REGIME:
  Yield curve         : {'inverted' if last_macro.get('yield_spread', 0) < 0 else 'normal'}
  Risk environment    : {'risk-off' if last_macro.get('vix', 20) > 25 else 'risk-on'}
""".strip()

        documents.append(Document(
            page_content=macro_doc,
            metadata={"type": "macro_context",
                      "date": str(macro_df_daily.index[-1].date())}
        ))

    if "news_df" in globals() and not news_df.empty:
        for _, row in news_df.head(30).iterrows():
            documents.append(Document(
                page_content=f"NEWS: {row.get('title', '')} — {row.get('summary', '')}",
                metadata={
                    "type":      "news",
                    "source":    row.get("source", ""),
                    "sentiment": row.get("sentiment_label", "neutral"),
                    "date":      str(row.get("date", "")),
                }
            ))

    print(f"Corpus built : {len(documents)} documents")
    return documents


corpus = build_financial_corpus(market_data, ticker_sentiment, macro_df_daily)

splitter = RecursiveCharacterTextSplitter(
    chunk_size=CONFIG["chunk_size"],
    chunk_overlap=CONFIG["chunk_overlap"],
    separators=["\n\n", "\n", " ", ""]
)
chunks = splitter.split_documents(corpus)
print(f"Total chunks : {len(chunks)}")

print("Building FAISS vector store...")
vectorstore = CustomVectorStore(embedder)
vectorstore.add_documents(chunks)
vectorstore.save(CONFIG["vectordb_dir"] / "faiss_main")
print("Vector store saved.")

hybrid_retriever = HybridRetriever(
    vectorstore=vectorstore,
    documents=chunks,
    alpha=0.5
)
print("Hybrid retriever ready.")

print("\nEmbedding benchmark...")
test_queries = [
    "What is the Sharpe ratio of AAPL?",
    "Which stock has the highest volatility?",
    "What is the current yield curve situation?",
    "Which ticker is overbought based on RSI?",
    "What is the VIX level and market risk?",
]

benchmark_results = []
for query in test_queries:
    dense_docs  = vectorstore.similarity_search(query, k=3)
    hybrid_docs = hybrid_retriever.retrieve(query, k=3)
    dense_rel   = sum(1 for d in dense_docs
                      if any(w in d.page_content.lower()
                             for w in query.lower().split())) / 3
    hybrid_rel  = sum(1 for d in hybrid_docs
                      if any(w in d.page_content.lower()
                             for w in query.lower().split())) / 3
    benchmark_results.append({
        "query":            query[:50],
        "dense_relevance":  dense_rel,
        "hybrid_relevance": hybrid_rel,
    })

print(f"\n{'Query':<52} {'Dense':>8} {'Hybrid':>8}")
print("-" * 72)
for r in benchmark_results:
    print(f"  {r['query']:<50} {r['dense_relevance']:>8.2f} {r['hybrid_relevance']:>8.2f}")

avg_dense  = np.mean([r["dense_relevance"]  for r in benchmark_results])
avg_hybrid = np.mean([r["hybrid_relevance"] for r in benchmark_results])
print(f"\n  Average Dense  : {avg_dense:.3f}")
print(f"  Average Hybrid : {avg_hybrid:.3f}")
print(f"  Hybrid gain    : +{(avg_hybrid - avg_dense)*100:.1f}%")

tracker.start("embedding_benchmark")
tracker.log_metric("avg_dense_relevance",  avg_dense)
tracker.log_metric("avg_hybrid_relevance", avg_hybrid)
tracker.log_param("embed_model",           CONFIG["embed_model"])
tracker.log_param("n_chunks",              len(chunks))
tracker.end()

print("\nBuilding knowledge graph...")
G = nx.Graph()

for ticker, df in market_data.items():
    returns = df["returns"].dropna()
    ann_ret = returns.mean() * 252 * 100
    ann_vol = returns.std() * np.sqrt(252) * 100
    sharpe  = (returns.mean() * 252) / (returns.std() * np.sqrt(252))
    trend   = "bullish" if df["ema_20"].iloc[-1] > df["ema_50"].iloc[-1] else "bearish"
    G.add_node(ticker, type="stock", ann_ret=ann_ret,
               ann_vol=ann_vol, sharpe=sharpe, trend=trend)

sectors = {
    "AAPL": "Technology", "MSFT": "Technology", "GOOGL": "Technology",
    "AMZN": "Consumer",   "NVDA": "Technology",  "META": "Technology",
    "TSLA": "Automotive", "JPM":  "Finance",      "GS":   "Finance",
    "AMD":  "Technology",
}
for sector in set(sectors.values()):
    G.add_node(sector, type="sector")
for ticker, sector in sectors.items():
    if ticker in G:
        G.add_edge(ticker, sector, relation="belongs_to")

returns_df  = pd.DataFrame({t: market_data[t]["returns"] for t in market_data}).dropna()
corr_matrix = returns_df.corr()
for i, t1 in enumerate(CONFIG["market_tickers"]):
    for t2 in CONFIG["market_tickers"][i+1:]:
        if t1 in corr_matrix.columns and t2 in corr_matrix.columns:
            corr = corr_matrix.loc[t1, t2]
            if abs(corr) > 0.6:
                G.add_edge(t1, t2, relation="correlated", weight=round(corr, 3))

for node in ["FED_RATE", "VIX", "YIELD_CURVE", "OIL", "GOLD"]:
    G.add_node(node, type="macro")
for ticker, macros in {"JPM": ["FED_RATE", "YIELD_CURVE"],
                        "GS":  ["FED_RATE", "YIELD_CURVE"],
                        "TSLA":["OIL"], "NVDA":["VIX"]}.items():
    for macro in macros:
        if ticker in G:
            G.add_edge(ticker, macro, relation="influenced_by")

print(f"  Nodes : {G.number_of_nodes()}")
print(f"  Edges : {G.number_of_edges()}")

with open(CONFIG["data_dir"] / "knowledge_graph.pkl", "wb") as f:
    pickle.dump(G, f)

print("\nGenerating t-SNE + PCA visualizations...")
sample_texts  = [c.page_content[:200] for c in chunks[:80]]
sample_labels = [c.metadata.get("ticker", c.metadata.get("type", "other"))
                 for c in chunks[:80]]
embeddings_matrix = embedder.embed(sample_texts)

tsne   = TSNE(n_components=2, random_state=42, perplexity=15)
emb_2d = tsne.fit_transform(embeddings_matrix)

unique_labels = list(set(sample_labels))
color_map     = {l: px.colors.qualitative.Plotly[i % 10]
                 for i, l in enumerate(unique_labels)}

fig = go.Figure()
for label in unique_labels:
    mask = np.array([l == label for l in sample_labels])
    fig.add_trace(go.Scatter(
        x=emb_2d[mask, 0], y=emb_2d[mask, 1],
        mode="markers", name=label,
        marker=dict(size=8, color=color_map[label]),
    ))
fig.update_layout(
    title="t-SNE — Financial Document Embeddings",
    height=600, template="plotly_white",
    xaxis_title="t-SNE 1", yaxis_title="t-SNE 2",
)
fig.show()
save_fig(fig, "tsne_embeddings.png")

pca     = PCA(n_components=2, random_state=42)
emb_pca = pca.fit_transform(embeddings_matrix)

fig2 = go.Figure()
for label in unique_labels:
    mask = np.array([l == label for l in sample_labels])
    fig2.add_trace(go.Scatter(
        x=emb_pca[mask, 0], y=emb_pca[mask, 1],
        mode="markers", name=label,
        marker=dict(size=8, color=color_map[label]),
    ))
fig2.update_layout(
    title="PCA — Financial Document Embeddings",
    height=600, template="plotly_white",
    xaxis_title="PC1", yaxis_title="PC2",
)
fig2.show()
save_fig(fig2, "pca_embeddings.png")

explained = pca.explained_variance_ratio_
print(f"PCA explained variance : PC1={explained[0]*100:.1f}%  PC2={explained[1]*100:.1f}%")

print()
print("=" * 65)
print("Cell 5 Summary")
print("=" * 65)
print(f"Corpus documents      : {len(corpus)}")
print(f"Total chunks          : {len(chunks)}")
print(f"Vector store          : {vectorstore.index.ntotal} vectors")
print(f"Hybrid retriever      : BM25 + FAISS (alpha=0.5)")
print(f"Knowledge graph       : {G.number_of_nodes()} nodes, {G.number_of_edges()} edges")
print(f"Embedding dim         : {embeddings_matrix.shape[1]}")
print(f"Avg dense relevance   : {avg_dense:.3f}")
print(f"Avg hybrid relevance  : {avg_hybrid:.3f}")
print()
print("Cell 5 done. Tell me and I give Cell 6.")
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_groq import ChatGroq
import time

llm = ChatGroq(
    model=CONFIG["llm_model"],
    api_key=os.environ["GROQ_API_KEY"],
    temperature=0.1,
    max_tokens=1024,
)

def format_docs(docs):
    return "\n\n---\n\n".join([
        f"[Source: {d.metadata.get('ticker', d.metadata.get('type', 'N/A'))} | "
        f"Type: {d.metadata.get('type', 'N/A')}]\n{d.page_content}"
        for d in docs
    ])

rag_prompt = ChatPromptTemplate.from_template("""
You are FinSight AI Pro, an expert quantitative financial analyst.
Use ONLY the following market data to answer the question.
Always cite specific numbers, percentages, and metrics from the context.

Context:
{context}

Question: {question}

Provide a detailed, data-driven analysis:""")

def rag_query_dense(question, k=5):
    docs    = vectorstore.similarity_search(question, k=k)
    context = format_docs(docs)
    chain   = rag_prompt | llm | StrOutputParser()
    answer  = chain.invoke({"context": context, "question": question})
    return answer, docs, context

def rag_query_hybrid(question, k=5):
    docs    = hybrid_retriever.retrieve(question, k=k)
    context = format_docs(docs)
    chain   = rag_prompt | llm | StrOutputParser()
    answer  = chain.invoke({"context": context, "question": question})
    return answer, docs, context

def generate_sub_queries(question, n=3):
    prompt = f"""
Generate {n} different reformulations of this financial question.
Each should focus on a different aspect to improve retrieval.
Return only the reformulations, one per line.

Original: {question}

Reformulations:"""
    response = call_llm(
        "You are a financial query reformulation expert.",
        prompt, max_tokens=200
    )
    sub_queries = [q.strip() for q in response.strip().split("\n")
                   if q.strip() and len(q.strip()) > 10]
    return sub_queries[:n]

def multi_query_rag(question, n_subqueries=2, k_per_query=3):
    sub_queries = generate_sub_queries(question, n=n_subqueries)
    all_docs    = {}
    for q in [question] + sub_queries:
        docs = hybrid_retriever.retrieve(q, k=k_per_query)
        for doc in docs:
            all_docs[doc.page_content[:100]] = doc
    unique_docs = list(all_docs.values())[:8]
    context     = format_docs(unique_docs)

    multi_prompt = ChatPromptTemplate.from_template("""
You are FinSight AI Pro, an expert quantitative financial analyst.
Use ONLY the provided context. Cite specific numbers and metrics.

Context:
{context}

Question: {question}

Sub-questions explored:
{sub_questions}

Provide a comprehensive analysis:""")

    chain  = multi_prompt | llm | StrOutputParser()
    answer = chain.invoke({
        "context":       context,
        "question":      question,
        "sub_questions": "\n".join([f"- {q}" for q in sub_queries]),
    })
    return answer, unique_docs, context, sub_queries

def hyde_rag(question, k=5):
    hyp_doc = call_llm(
        "You are a financial document writer.",
        f"Write a 100-word hypothetical financial analysis that answers: {question}",
        max_tokens=150
    )
    hyp_emb = embedder.embed_query(hyp_doc)
    hyp_emb = np.array([hyp_emb], dtype=np.float32)
    import faiss
    faiss.normalize_L2(hyp_emb)
    scores, indices = vectorstore.index.search(hyp_emb, k)
    docs    = [vectorstore.documents[i] for i in indices[0]
               if i < len(vectorstore.documents)]
    context = format_docs(docs)
    chain   = rag_prompt | llm | StrOutputParser()
    answer  = chain.invoke({"context": context, "question": question})
    return answer, docs, context

eval_questions = [
    "Which ticker has the highest Sharpe ratio and what are the implications?",
    "Compare AAPL and NVDA in terms of volatility, trend and risk.",
    "Which stocks are overbought or oversold based on RSI?",
    "What is the maximum drawdown of TSLA vs other tickers?",
    "What is the current macro environment and how does it affect the portfolio?",
    "Which stocks show golden cross signals and what is their momentum?",
]

print("=" * 70)
print("Advanced RAG Evaluation — FinSight AI Pro")
print("=" * 70)
print()

rag_eval_results = []

for i, question in enumerate(eval_questions, 1):
    print(f"Question {i}/{len(eval_questions)}: {question[:65]}")
    print("-" * 55)

    time.sleep(5)
    t0                               = time.time()
    ans_dense, docs_dense, ctx_dense = rag_query_dense(question, k=4)
    t_dense                          = time.time() - t0

    time.sleep(5)
    t0                                 = time.time()
    ans_hybrid, docs_hybrid, ctx_hybrid = rag_query_hybrid(question, k=4)
    t_hybrid                           = time.time() - t0

    time.sleep(5)
    t0                                       = time.time()
    ans_multi, docs_multi, ctx_multi, sub_qs = multi_query_rag(
        question, n_subqueries=2
    )
    t_multi                                  = time.time() - t0

    eval_dense  = rag_evaluator.evaluate(question, ctx_dense,  ans_dense)
    eval_hybrid = rag_evaluator.evaluate(question, ctx_hybrid, ans_hybrid)
    eval_multi  = rag_evaluator.evaluate(question, ctx_multi,  ans_multi)

    best = max(
        [("Dense",   ans_dense,  eval_dense),
         ("Hybrid",  ans_hybrid, eval_hybrid),
         ("Multi-Q", ans_multi,  eval_multi)],
        key=lambda x: x[2]["overall"]
    )

    print(f"Best method  : {best[0]} (score={best[2]['overall']:.3f})")
    print(f"Answer       : {best[1][:200]}...")
    print(f"  Dense      : score={eval_dense['overall']:.3f}  "
          f"faith={eval_dense['faithfulness']:.3f}  lat={t_dense:.1f}s")
    print(f"  Hybrid     : score={eval_hybrid['overall']:.3f}  "
          f"faith={eval_hybrid['faithfulness']:.3f}  lat={t_hybrid:.1f}s")
    print(f"  Multi-Q    : score={eval_multi['overall']:.3f}  "
          f"faith={eval_multi['faithfulness']:.3f}  lat={t_multi:.1f}s")
    print()

    rag_eval_results.append({
        "question":       question,
        "best_method":    best[0],
        "best_score":     best[2]["overall"],
        "dense_score":    eval_dense["overall"],
        "hybrid_score":   eval_hybrid["overall"],
        "multi_score":    eval_multi["overall"],
        "dense_latency":  t_dense,
        "hybrid_latency": t_hybrid,
        "multi_latency":  t_multi,
        "sub_queries":    sub_qs,
        "best_answer":    best[1],
    })

    time.sleep(8)

print("=" * 70)
print("RAG Evaluation Summary")
print("=" * 70)
for method, key in [("Dense",   "dense_score"),
                    ("Hybrid",  "hybrid_score"),
                    ("Multi-Q", "multi_score")]:
    avg     = np.mean([r[key] for r in rag_eval_results])
    lat_key = key.replace("score", "latency")
    avg_lat = np.mean([r[lat_key] for r in rag_eval_results])
    print(f"  {method:<12} : avg_score={avg:.3f}  avg_latency={avg_lat:.1f}s")

best_counts = {}
for r in rag_eval_results:
    best_counts[r["best_method"]] = best_counts.get(r["best_method"], 0) + 1
print()
print("Best method per question :")
for method, count in sorted(best_counts.items(), key=lambda x: -x[1]):
    print(f"  {method:<12} : {count}/{len(rag_eval_results)}")

tracker.start("rag_evaluation")
tracker.log_metric("avg_dense_score",  np.mean([r["dense_score"]  for r in rag_eval_results]))
tracker.log_metric("avg_hybrid_score", np.mean([r["hybrid_score"] for r in rag_eval_results]))
tracker.log_metric("avg_multi_score",  np.mean([r["multi_score"]  for r in rag_eval_results]))
tracker.log_param("n_questions",       len(rag_eval_results))
tracker.end()

colors_rag = ["#1F77B4", "#FF7F0E", "#2CA02C"]
all_scores  = {
    "Dense":   [r["dense_score"]  for r in rag_eval_results],
    "Hybrid":  [r["hybrid_score"] for r in rag_eval_results],
    "Multi-Q": [r["multi_score"]  for r in rag_eval_results],
}

fig = go.Figure()
for j, (method, scores) in enumerate(all_scores.items()):
    fig.add_trace(go.Bar(
        name=method,
        x=[f"Q{i+1}" for i in range(len(scores))],
        y=scores,
        marker_color=colors_rag[j],
    ))
fig.update_layout(
    title="RAG Evaluation — Score by Question and Method",
    barmode="group",
    yaxis=dict(range=[0, 1.1]),
    height=450, template="plotly_white",
)
fig.show()
save_fig(fig, "rag_comparison.png")

fig2 = go.Figure()
for j, (method, scores) in enumerate(all_scores.items()):
    fig2.add_trace(go.Box(
        y=scores, name=method,
        marker_color=colors_rag[j],
        boxpoints="all",
    ))
fig2.update_layout(
    title="RAG Score Distribution by Method",
    yaxis=dict(range=[0, 1.1]),
    height=400, template="plotly_white",
)
fig2.show()
save_fig(fig2, "rag_score_distribution.png")

latencies = {
    "Dense":   [r["dense_latency"]  for r in rag_eval_results],
    "Hybrid":  [r["hybrid_latency"] for r in rag_eval_results],
    "Multi-Q": [r["multi_latency"]  for r in rag_eval_results],
}
fig3 = go.Figure()
for j, (method, lats) in enumerate(latencies.items()):
    fig3.add_trace(go.Scatter(
        x=[f"Q{i+1}" for i in range(len(lats))],
        y=lats,
        mode="lines+markers",
        name=method,
        line=dict(color=colors_rag[j], width=2),
    ))
fig3.update_layout(
    title="RAG Latency by Question and Method (seconds)",
    yaxis_title="Latency (s)",
    height=400, template="plotly_white",
)
fig3.show()
save_fig(fig3, "rag_latency.png")

with open(CONFIG["results_dir"] / "rag_evaluation.json", "w") as f:
    json.dump(rag_eval_results, f, indent=2, default=str)

print()
print("=" * 70)
print("Cell 6 Summary")
print("=" * 70)
print(f"Questions evaluated    : {len(rag_eval_results)}")
print(f"RAG techniques         : Dense, Hybrid, Multi-Query, HyDE")
print(f"Avg Dense score        : {np.mean([r['dense_score'] for r in rag_eval_results]):.3f}")
print(f"Avg Hybrid score       : {np.mean([r['hybrid_score'] for r in rag_eval_results]):.3f}")
print(f"Avg Multi-Q score      : {np.mean([r['multi_score'] for r in rag_eval_results]):.3f}")
print()
print("Cell 6 done. Tell me and I give Cell 7.")
```

```python
# Cell 7 — Multi-Agent System avec LangGraph
# 10 agents specialises + orchestrateur

from langgraph.graph import StateGraph, END
from typing import TypedDict, List, Dict, Any, Optional

class AgentState(TypedDict):
    question:       str
    ticker:         str
    fundamental:    str
    technical:      str
    sentiment:      str
    macro:          str
    risk:           str
    regime:         str
    news:           str
    rag_context:    str
    final_report:   str
    recommendation: str
    confidence:     float
    price_target:   str
    messages:       List[str]

# ------------------------------------------------------------
# Agent 1 : Analyste Fondamental
# ------------------------------------------------------------

def fundamental_agent(state: AgentState) -> AgentState:
    ticker = state["ticker"]
    df     = market_data.get(ticker)
    if df is None:
        state["fundamental"] = f"No data for {ticker}"
        return state

    returns = df["returns"].dropna()
    last    = df.iloc[-1]
    ann_ret = returns.mean() * 252 * 100
    ann_vol = returns.std() * np.sqrt(252) * 100
    sharpe  = (returns.mean() * 252) / (returns.std() * np.sqrt(252))
    max_dd  = ((df["Close"] / df["Close"].cummax()) - 1).min() * 100
    calmar  = ann_ret / abs(max_dd)
    ytd     = (df["Close"].iloc[-1] / df["Close"].iloc[0] - 1) * 100
    var_95  = np.percentile(returns, 5) * 100
    cvar_95 = returns[returns <= np.percentile(returns, 5)].mean() * 100

    prompt = f"""
Analyze {ticker} fundamentals:
- Current Price    : ${last['Close']:.2f}
- YTD Return       : {ytd:.2f}%
- Annual Return    : {ann_ret:.2f}%
- Annual Vol       : {ann_vol:.2f}%
- Sharpe Ratio     : {sharpe:.3f}
- Calmar Ratio     : {calmar:.3f}
- Max Drawdown     : {max_dd:.2f}%
- VaR 95%          : {var_95:.2f}%
- CVaR 95%         : {cvar_95:.2f}%
- 52W High         : ${df['Close'].tail(252).max():.2f}
- 52W Low          : ${df['Close'].tail(252).min():.2f}
- Avg Volume (30d) : {df['Volume'].tail(30).mean():,.0f}

Provide a 3-sentence fundamental assessment focusing on
risk-adjusted performance and investor suitability.
"""
    state["fundamental"] = call_llm(
        "You are a senior fundamental analyst. Be concise and data-driven.",
        prompt, max_tokens=250
    )
    state["messages"].append(f"[Fundamental] Analyzed {ticker}")
    return state

# ------------------------------------------------------------
# Agent 2 : Analyste Technique
# ------------------------------------------------------------

def technical_agent(state: AgentState) -> AgentState:
    ticker = state["ticker"]
    df     = market_data.get(ticker)
    if df is None:
        state["technical"] = f"No data for {ticker}"
        return state

    last      = df.iloc[-1]
    trend     = "BULLISH" if last["ema_20"] > last["ema_50"] else "BEARISH"
    lt_trend  = "BULLISH" if last["Close"] > last["ema_200"] else "BEARISH"
    rsi_sig   = "OVERBOUGHT" if last["rsi"] > 70 else "OVERSOLD" if last["rsi"] < 30 else "NEUTRAL"
    bb_pos    = "ABOVE UPPER" if last["Close"] > last["bb_high"] \
        else "BELOW LOWER" if last["Close"] < last["bb_low"] \
        else "WITHIN BANDS"
    macd_dir  = "BULLISH" if last["macd_diff"] > 0 else "BEARISH"

    prompt = f"""
Technical analysis for {ticker}:
- Price        : ${last['Close']:.2f}
- Short trend  : {trend} (EMA20={last['ema_20']:.2f} vs EMA50={last['ema_50']:.2f})
- Long trend   : {lt_trend} (vs EMA200={last['ema_200']:.2f})
- RSI (14)     : {last['rsi']:.2f} — {rsi_sig}
- MACD         : {last['macd']:.4f} ({macd_dir})
- Bollinger    : Price {bb_pos}
- ADX          : {last['adx']:.2f}
- Stochastic   : {last['stoch']:.2f}
- ATR          : {last['atr']:.2f}
- Vol Ratio    : {last['vol_ratio']:.2f}x
- Golden Cross : {'YES' if last['golden_cross'] else 'NO'}
- Fib 38.2%    : ${last['fib_382']:.2f}
- Fib 61.8%    : ${last['fib_618']:.2f}

Provide a 3-sentence technical assessment covering
trend, momentum, and key support/resistance.
"""
    state["technical"] = call_llm(
        "You are a senior technical analyst. Be concise and precise.",
        prompt, max_tokens=250
    )
    state["messages"].append(f"[Technical] Analyzed {ticker}")
    return state

# ------------------------------------------------------------
# Agent 3 : Analyste Sentiment
# ------------------------------------------------------------

def sentiment_agent(state: AgentState) -> AgentState:
    ticker      = state["ticker"]
    df          = market_data.get(ticker)
    sent        = ticker_sentiment.get(ticker, {})

    if df is None:
        state["sentiment"] = f"No data for {ticker}"
        return state

    recent      = df["returns"].tail(20) * 100
    positive_d  = (recent > 0).sum()
    negative_d  = (recent < 0).sum()
    obv_trend   = "INCREASING" if df["obv"].iloc[-1] > df["obv"].iloc[-20] else "DECREASING"
    vol_ratio   = df["vol_ratio"].iloc[-1]
    mfi         = df.iloc[-1]["mfi"]

    prompt = f"""
Sentiment analysis for {ticker}:
- News Score      : {sent.get('avg_score', 0):.3f} ({sent.get('label', 'neutral')})
- Positive news   : {sent.get('positive_pct', 0):.1f}%
- Negative news   : {sent.get('negative_pct', 0):.1f}%
- Positive days   : {positive_d}/20
- Negative days   : {negative_d}/20
- Avg 20d return  : {recent.mean():.3f}%
- OBV trend       : {obv_trend}
- Volume ratio    : {vol_ratio:.2f}x
- MFI             : {mfi:.2f}

Provide a 2-sentence sentiment assessment covering
market mood and institutional interest.
"""
    state["sentiment"] = call_llm(
        "You are a market sentiment analyst. Be concise.",
        prompt, max_tokens=150
    )
    state["messages"].append(f"[Sentiment] Analyzed {ticker}")
    return state

# ------------------------------------------------------------
# Agent 4 : Analyste Macro
# ------------------------------------------------------------

def macro_agent(state: AgentState) -> AgentState:
    last_macro = macro_df_daily.iloc[-1]
    vix        = last_macro.get("vix", 18)
    spread     = last_macro.get("yield_spread", 0.5)
    fed_rate   = last_macro.get("fed_rate", 2.0)
    cpi_yoy    = last_macro.get("cpi_yoy", 2.5)
    credit     = last_macro.get("credit_spread", 4.0)

    regime = "risk-off" if vix > 25 else "risk-on"
    curve  = "inverted" if spread < 0 else "normal"

    tickers_summary = []
    for t, df in market_data.items():
        ret  = df["returns"].mean() * 252 * 100
        vol  = df["returns"].std() * np.sqrt(252) * 100
        corr = df["returns"].corr(market_data["AAPL"]["returns"])
        tickers_summary.append(f"{t}: ret={ret:.1f}%, vol={vol:.1f}%, corr_AAPL={corr:.2f}")

    prompt = f"""
Macro context:
- VIX           : {vix:.2f} ({regime})
- Yield curve   : {curve} (spread={spread:.2f}%)
- Fed rate      : {fed_rate:.2f}%
- CPI YoY       : {cpi_yoy:.2f}%
- Credit spread : {credit:.2f}%
- Oil           : ${last_macro.get('oil_wti', 70):.2f}/barrel
- Gold          : ${last_macro.get('gold', 1800):.2f}/oz

Portfolio: {', '.join(tickers_summary[:5])}

Assess in 3 sentences: market regime, sector risk, and portfolio impact.
"""
    state["macro"] = call_llm(
        "You are a senior macro analyst.",
        prompt, max_tokens=200
    )
    state["messages"].append("[Macro] Market regime assessed")
    return state

# ------------------------------------------------------------
# Agent 5 : Analyste Risque
# ------------------------------------------------------------

def risk_agent(state: AgentState) -> AgentState:
    ticker = state["ticker"]
    df     = market_data.get(ticker)
    if df is None:
        state["risk"] = f"No data for {ticker}"
        return state

    returns = df["returns"].dropna()
    var_95  = np.percentile(returns, 5) * 100
    var_99  = np.percentile(returns, 1) * 100
    cvar_95 = returns[returns <= np.percentile(returns, 5)].mean() * 100
    max_dd  = ((df["Close"] / df["Close"].cummax()) - 1).min() * 100
    skew    = returns.skew()
    kurt    = returns.kurtosis()
    ann_vol = returns.std() * np.sqrt(252) * 100

    prompt = f"""
Risk analysis for {ticker}:
- VaR 95%         : {var_95:.2f}%
- VaR 99%         : {var_99:.2f}%
- CVaR 95%        : {cvar_95:.2f}%
- Max Drawdown    : {max_dd:.2f}%
- Annual Vol      : {ann_vol:.2f}%
- Return Skew     : {skew:.3f}
- Return Kurtosis : {kurt:.3f}

Provide a 2-sentence risk assessment and position sizing recommendation.
"""
    state["risk"] = call_llm(
        "You are a quantitative risk manager.",
        prompt, max_tokens=150
    )
    state["messages"].append(f"[Risk] Analyzed {ticker}")
    return state

# ------------------------------------------------------------
# Agent 6 : Detecteur de Regime de Marche (HMM)
# ------------------------------------------------------------

def regime_agent(state: AgentState) -> AgentState:
    ticker = state["ticker"]
    df     = market_data.get(ticker)
    if df is None:
        state["regime"] = "unknown"
        return state

    try:
        features = df[["returns", "vol_20", "rsi", "macd"]].dropna().tail(500)
        features_scaled = StandardScaler().fit_transform(features)

        model = hmm.GaussianHMM(
            n_components=3, covariance_type="full",
            n_iter=100, random_state=42
        )
        model.fit(features_scaled)
        hidden_states = model.predict(features_scaled)

        current_state = hidden_states[-1]
        state_returns = {}
        for s in range(3):
            mask = hidden_states == s
            if mask.sum() > 0:
                state_returns[s] = features["returns"].values[mask].mean() * 252 * 100

        sorted_states = sorted(state_returns.items(), key=lambda x: x[1])
        regime_map    = {
            sorted_states[0][0]: "BEAR",
            sorted_states[1][0]: "SIDEWAYS",
            sorted_states[2][0]: "BULL",
        }
        current_regime = regime_map.get(current_state, "UNKNOWN")

        state_durations = {}
        for s in range(3):
            state_durations[regime_map[s]] = int((hidden_states == s).sum())

        state["regime"] = (
            f"Current regime : {current_regime} | "
            f"Bull days: {state_durations.get('BULL', 0)} | "
            f"Bear days: {state_durations.get('BEAR', 0)} | "
            f"Sideways days: {state_durations.get('SIDEWAYS', 0)}"
        )
    except Exception as e:
        state["regime"] = f"Regime detection error : {str(e)[:50]}"

    state["messages"].append(f"[Regime] Detected for {ticker}")
    return state

# ------------------------------------------------------------
# Agent 7 : Anomaly Detector
# ------------------------------------------------------------

def anomaly_agent(state: AgentState) -> AgentState:
    ticker = state["ticker"]
    df     = market_data.get(ticker)
    if df is None:
        state["messages"].append(f"[Anomaly] No data for {ticker}")
        return state

    features = df[["returns", "vol_20", "vol_ratio", "rsi", "atr"]].dropna().tail(252)
    iso      = IsolationForest(contamination=0.05, random_state=42)
    labels   = iso.fit_predict(features)
    n_anomaly = (labels == -1).sum()

    recent_anomalies = (labels[-20:] == -1).sum()
    latest_return    = df["returns"].iloc[-1] * 100
    z_score          = (latest_return - df["returns"].mean() * 100) / \
                       (df["returns"].std() * 100)

    anomaly_info = (
        f"Anomalies (252d): {n_anomaly} ({n_anomaly/252*100:.1f}%) | "
        f"Recent (20d): {recent_anomalies} | "
        f"Latest return z-score: {z_score:.2f} | "
        f"Alert: {'YES — unusual activity' if abs(z_score) > 2 else 'NO'}"
    )
    state["messages"].append(f"[Anomaly] {ticker} — {anomaly_info}")
    return state

# ------------------------------------------------------------
# Agent 8 : News Agent
# ------------------------------------------------------------

def news_agent(state: AgentState) -> AgentState:
    ticker      = state["ticker"]
    ticker_news = news_df[
        news_df["tickers_mentioned"].apply(
            lambda x: ticker in x if isinstance(x, list) else False
        )
    ].head(5)

    if ticker_news.empty:
        ticker_news = news_df.sample(min(3, len(news_df)))

    news_summary = "\n".join([
        f"- [{row['sentiment_label'].upper()}] {row['title'][:80]}"
        for _, row in ticker_news.iterrows()
    ])

    state["news"] = f"Recent news for {ticker}:\n{news_summary}"
    state["messages"].append(f"[News] Retrieved {len(ticker_news)} articles for {ticker}")
    return state

# ------------------------------------------------------------
# Agent 9 : RAG Agent
# ------------------------------------------------------------

def rag_agent(state: AgentState) -> AgentState:
    ticker  = state["ticker"]
    query   = f"complete analysis of {ticker} including technicals, fundamentals, risk and performance"
    docs    = hybrid_retriever.retrieve(query, k=CONFIG["top_k"])
    context = format_docs(docs)
    state["rag_context"] = context
    state["messages"].append(f"[RAG] Retrieved {len(docs)} docs for {ticker}")
    return state

# ------------------------------------------------------------
# Agent 10 : Orchestrateur CIO
# ------------------------------------------------------------

def orchestrator_agent(state: AgentState) -> AgentState:
    ticker = state["ticker"]
    df     = market_data.get(ticker)
    last   = df.iloc[-1] if df is not None else {}
    price  = last.get("Close", 0) if hasattr(last, "get") else 0

    prompt = f"""
You are the Chief Investment Officer of FinSight AI Pro.
Synthesize ALL analyses below into a definitive investment report for {ticker}.

FUNDAMENTAL ANALYSIS:
{state['fundamental']}

TECHNICAL ANALYSIS:
{state['technical']}

SENTIMENT ANALYSIS:
{state['sentiment']}

MACRO CONTEXT:
{state['macro']}

RISK ANALYSIS:
{state['risk']}

MARKET REGIME:
{state['regime']}

NEWS:
{state['news']}

RAG MARKET DATA:
{state['rag_context'][:400]}

Current price: ${price:.2f}

Deliver:
1. Executive Summary (2 sentences)
2. Key Strengths (2 bullet points with specific metrics)
3. Key Risks (2 bullet points with specific metrics)
4. Recommendation: BUY / HOLD / SELL
5. Confidence Level: X%
6. Price Target Range: $X — $Y (6-month horizon)
"""
    report = call_llm(
        "You are a Chief Investment Officer. Be decisive and data-driven.",
        prompt, max_tokens=600
    )

    rec  = "HOLD"
    conf = 0.5
    upper = report.upper()
    if "STRONG BUY"  in upper: rec = "STRONG BUY";  conf = 0.90
    elif "BUY"       in upper: rec = "BUY";          conf = 0.75
    elif "STRONG SELL" in upper: rec = "STRONG SELL"; conf = 0.85
    elif "SELL"      in upper: rec = "SELL";         conf = 0.70

    state["final_report"]   = report
    state["recommendation"] = rec
    state["confidence"]     = conf
    state["messages"].append(
        f"[Orchestrator] Report generated — {rec} ({conf*100:.0f}%)"
    )
    return state

# ------------------------------------------------------------
# Construction du graphe LangGraph
# ------------------------------------------------------------

def build_agent_graph():
    graph = StateGraph(AgentState)

    graph.add_node("fundamental",  fundamental_agent)
    graph.add_node("technical",    technical_agent)
    graph.add_node("sentiment",    sentiment_agent)
    graph.add_node("macro",        macro_agent)
    graph.add_node("risk",         risk_agent)
    graph.add_node("regime",       regime_agent)
    graph.add_node("anomaly",      anomaly_agent)
    graph.add_node("news",         news_agent)
    graph.add_node("rag",          rag_agent)
    graph.add_node("orchestrator", orchestrator_agent)

    graph.set_entry_point("fundamental")
    graph.add_edge("fundamental", "technical")
    graph.add_edge("technical",   "sentiment")
    graph.add_edge("sentiment",   "macro")
    graph.add_edge("macro",       "risk")
    graph.add_edge("risk",        "regime")
    graph.add_edge("regime",      "anomaly")
    graph.add_edge("anomaly",     "news")
    graph.add_edge("news",        "rag")
    graph.add_edge("rag",         "orchestrator")
    graph.add_edge("orchestrator", END)

    return graph.compile()

agent_graph = build_agent_graph()
print("Multi-agent graph compiled : 10 agents + orchestrator")
print()

# ------------------------------------------------------------
# Execution
# ------------------------------------------------------------

def run_agent_analysis(ticker):
    print(f"Running 10-agent analysis for {ticker}...")

    initial_state = AgentState(
        question=       f"Complete investment analysis for {ticker}",
        ticker=         ticker,
        fundamental=    "",
        technical=      "",
        sentiment=      "",
        macro=          "",
        risk=           "",
        regime=         "",
        news=           "",
        rag_context=    "",
        final_report=   "",
        recommendation= "",
        confidence=     0.0,
        price_target=   "",
        messages=       [],
    )

    result = agent_graph.invoke(initial_state)

    print("=" * 65)
    print(f"FinSight AI Pro — Investment Report : {ticker}")
    print("=" * 65)
    print(result["final_report"])
    print()
    print(f"Recommendation : {result['recommendation']}")
    print(f"Confidence     : {result['confidence']*100:.0f}%")
    print(f"Regime         : {result['regime']}")
    print()
    print("Agent pipeline :")
    for msg in result["messages"]:
        print(f"  {msg}")
    print()

    return result

tickers_to_analyze = ["AAPL", "NVDA", "TSLA"]
agent_results      = {}

for ticker in tickers_to_analyze:
    time.sleep(5)
    agent_results[ticker] = run_agent_analysis(ticker)
    print("-" * 65)
    print()

# ------------------------------------------------------------
# Visualisation des recommandations
# ------------------------------------------------------------

rec_map    = {"STRONG BUY": 2, "BUY": 1, "HOLD": 0, "SELL": -1, "STRONG SELL": -2}
tickers_r  = list(agent_results.keys())
recs       = [agent_results[t]["recommendation"] for t in tickers_r]
confs      = [agent_results[t]["confidence"] * 100 for t in tickers_r]
rec_values = [rec_map.get(r, 0) for r in recs]

colors_rec = []
for r in recs:
    if r in ["STRONG BUY", "BUY"]:
        colors_rec.append("#26A69A")
    elif r in ["STRONG SELL", "SELL"]:
        colors_rec.append("#EF5350")
    else:
        colors_rec.append("#FFA726")

fig = make_subplots(
    rows=1, cols=2,
    subplot_titles=("Agent Recommendations", "Confidence Level (%)")
)
fig.add_trace(go.Bar(
    x=tickers_r, y=rec_values,
    marker_color=colors_rec,
    text=recs, textposition="outside",
    name="Recommendation"
), row=1, col=1)
fig.add_trace(go.Bar(
    x=tickers_r, y=confs,
    marker_color="#1F77B4",
    text=[f"{c:.0f}%" for c in confs],
    textposition="outside",
    name="Confidence"
), row=1, col=2)
fig.update_layout(
    title="FinSight AI Pro — 10-Agent Investment Recommendations",
    height=450, template="plotly_white", showlegend=False,
)
fig.update_yaxes(title_text="Signal (-2=Strong Sell, 2=Strong Buy)", row=1, col=1)
fig.update_yaxes(title_text="Confidence (%)", row=1, col=2)
fig.show()
save_fig(fig, "agent_recommendations.png")

tracker.start("multi_agent_analysis")
tracker.log_param("tickers_analyzed", str(tickers_to_analyze))
tracker.log_param("n_agents",         10)
for t in tickers_to_analyze:
    tracker.log_metric(f"confidence_{t}", agent_results[t]["confidence"])
tracker.end()

print()
print("=" * 65)
print("Cell 7 Summary")
print("=" * 65)
print(f"Agents deployed    : 10 (Fundamental, Technical, Sentiment,")
print(f"                        Macro, Risk, Regime, Anomaly, News,")
print(f"                        RAG, Orchestrator)")
print(f"Tickers analyzed   : {tickers_to_analyze}")
for t in tickers_to_analyze:
    print(f"  {t:<6} : {agent_results[t]['recommendation']:<12} "
          f"({agent_results[t]['confidence']*100:.0f}% confidence)")
print()
print("Cell 7 done. Tell me and I give Cell 8.")
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_squared_error, mean_absolute_error
import numpy as np
import pickle

class StockDataset(Dataset):
    def __init__(self, X, y):
        self.X = torch.FloatTensor(X)
        self.y = torch.FloatTensor(y)

    def __len__(self):
        return len(self.X)

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]


def prepare_sequences(df, features, target_col, seq_len):
    data   = df[features + [target_col]].dropna().values
    scaler = MinMaxScaler()
    scaled = scaler.fit_transform(data)
    X, y   = [], []
    for i in range(seq_len, len(scaled)):
        X.append(scaled[i - seq_len:i, :-1])
        y.append(scaled[i, -1])
    return np.array(X), np.array(y), scaler


class LSTMAttention(nn.Module):
    def __init__(self, input_size, hidden_size=128,
                 num_layers=2, dropout=0.2):
        super().__init__()
        self.hidden_size = hidden_size
        self.num_layers  = num_layers

        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_layers,
            dropout=dropout,
            batch_first=True,
        )
        self.attention = nn.Sequential(
            nn.Linear(hidden_size, hidden_size // 2),
            nn.Tanh(),
            nn.Linear(hidden_size // 2, 1),
            nn.Softmax(dim=1),
        )
        self.fc = nn.Sequential(
            nn.Linear(hidden_size, hidden_size // 2),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_size // 2, 1),
        )

    def forward(self, x):
        h0 = torch.zeros(self.num_layers, x.size(0),
                         self.hidden_size).to(x.device)
        c0 = torch.zeros(self.num_layers, x.size(0),
                         self.hidden_size).to(x.device)
        out, _       = self.lstm(x, (h0, c0))
        attn_weights = self.attention(out)
        context      = torch.sum(attn_weights * out, dim=1)
        return self.fc(context).squeeze(-1)


class BiLSTMAttention(nn.Module):
    def __init__(self, input_size, hidden_size=128,
                 num_layers=2, dropout=0.2):
        super().__init__()
        self.hidden_size = hidden_size
        self.num_layers  = num_layers

        self.bilstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_layers,
            dropout=dropout,
            batch_first=True,
            bidirectional=True,
        )
        self.attention = nn.Sequential(
            nn.Linear(hidden_size * 2, hidden_size),
            nn.Tanh(),
            nn.Linear(hidden_size, 1),
            nn.Softmax(dim=1),
        )
        self.fc = nn.Sequential(
            nn.Linear(hidden_size * 2, hidden_size),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_size, 1),
        )

    def forward(self, x):
        out, _       = self.bilstm(x)
        attn_weights = self.attention(out)
        context      = torch.sum(attn_weights * out, dim=1)
        return self.fc(context).squeeze(-1)


def train_model(model, train_dl, val_dl, epochs, patience, lr):
    optimizer = optim.Adam(model.parameters(), lr=lr)
    scheduler = optim.lr_scheduler.ReduceLROnPlateau(
        optimizer, patience=5, factor=0.5
    )
    criterion     = nn.MSELoss()
    best_val_loss = float("inf")
    best_state    = None
    train_losses  = []
    val_losses    = []
    wait          = 0

    for epoch in range(epochs):
        model.train()
        train_loss = 0
        for xb, yb in train_dl:
            xb, yb = xb.to(CONFIG["device"]), yb.to(CONFIG["device"])
            optimizer.zero_grad()
            pred = model(xb)
            loss = criterion(pred, yb)
            loss.backward()
            nn.utils.clip_grad_norm_(model.parameters(), 1.0)
            optimizer.step()
            train_loss += loss.item()
        train_loss /= len(train_dl)

        model.eval()
        val_loss = 0
        with torch.no_grad():
            for xb, yb in val_dl:
                xb, yb   = xb.to(CONFIG["device"]), yb.to(CONFIG["device"])
                val_loss += criterion(model(xb), yb).item()
        val_loss /= len(val_dl)

        train_losses.append(train_loss)
        val_losses.append(val_loss)
        scheduler.step(val_loss)

        if val_loss < best_val_loss:
            best_val_loss = val_loss
            best_state    = {k: v.clone() for k, v in model.state_dict().items()}
            wait          = 0
        else:
            wait += 1
            if wait >= patience:
                print(f"    Early stopping at epoch {epoch+1}")
                break

        if (epoch + 1) % 20 == 0:
            print(f"    Epoch {epoch+1:3d} | Train: {train_loss:.6f} | Val: {val_loss:.6f}")

    model.load_state_dict(best_state)
    return model, train_losses, val_losses


def evaluate_model(model, test_dl, scaler, n_features):
    model.eval()
    preds, actuals = [], []
    with torch.no_grad():
        for xb, yb in test_dl:
            xb   = xb.to(CONFIG["device"])
            pred = model(xb).cpu().numpy()
            preds.extend(pred)
            actuals.extend(yb.numpy())

    preds   = np.array(preds)
    actuals = np.array(actuals)

    dummy_preds            = np.zeros((len(preds),   n_features))
    dummy_acts             = np.zeros((len(actuals), n_features))
    dummy_preds[:, -1]     = preds
    dummy_acts[:,  -1]     = actuals
    preds_inv   = scaler.inverse_transform(dummy_preds)[:, -1]
    actuals_inv = scaler.inverse_transform(dummy_acts)[:,  -1]

    rmse = np.sqrt(mean_squared_error(actuals_inv, preds_inv))
    mae  = mean_absolute_error(actuals_inv, preds_inv)
    mape = np.mean(np.abs((actuals_inv - preds_inv) / actuals_inv)) * 100
    corr = np.corrcoef(actuals_inv, preds_inv)[0, 1]

    return preds_inv, actuals_inv, rmse, mae, mape, corr


def train_lstm_pipeline(ticker, epochs=80, patience=15):
    df = market_data[ticker]

    base_features = [
        "Open", "High", "Low", "Close", "Volume",
        "rsi", "macd", "macd_sig", "macd_diff",
        "bb_high", "bb_low", "bb_mid", "bb_width",
        "atr", "ema_20", "ema_50", "ema_200",
        "obv", "mfi", "stoch", "williams", "roc",
        "returns", "volatility", "vol_ratio",
        "price_range", "gap", "momentum_20",
        "golden_cross", "adx", "cmf",
    ]
    macro_features = [c for c in df.columns if c.startswith("macro_")]
    features       = [f for f in base_features + macro_features if f in df.columns]
    target         = "target_5d"

    X, y, scaler = prepare_sequences(df, features, target, CONFIG["sequence_length"])

    n       = len(X)
    n_train = int(n * 0.70)
    n_val   = int(n * 0.15)

    X_train, y_train = X[:n_train],                y[:n_train]
    X_val,   y_val   = X[n_train:n_train + n_val], y[n_train:n_train + n_val]
    X_test,  y_test  = X[n_train + n_val:],        y[n_train + n_val:]

    train_dl = DataLoader(StockDataset(X_train, y_train),
                          batch_size=CONFIG["batch_size"], shuffle=True)
    val_dl   = DataLoader(StockDataset(X_val,   y_val),
                          batch_size=CONFIG["batch_size"])
    test_dl  = DataLoader(StockDataset(X_test,  y_test),
                          batch_size=CONFIG["batch_size"])

    results = {}

    for arch_name, ModelClass in [
        ("LSTM+Attention",   LSTMAttention),
        ("BiLSTM+Attention", BiLSTMAttention),
    ]:
        print(f"  Training {arch_name} for {ticker}...")
        model = ModelClass(
            input_size=len(features),
            hidden_size=128,
            num_layers=2,
            dropout=0.2,
        ).to(CONFIG["device"])

        model, train_losses, val_losses = train_model(
            model, train_dl, val_dl,
            epochs=epochs, patience=patience,
            lr=CONFIG["learning_rate"]
        )

        preds_inv, actuals_inv, rmse, mae, mape, corr = evaluate_model(
            model, test_dl, scaler, len(features) + 1
        )

        print(f"    RMSE: {rmse:.4f} | MAE: {mae:.4f} | "
              f"MAPE: {mape:.2f}% | Corr: {corr:.4f}")

        torch.save(
            model.state_dict(),
            CONFIG["models_dir"] / f"{arch_name.replace('+','_')}_{ticker}.pt"
        )

        tracker.start(f"{arch_name}_{ticker}")
        tracker.log_param("ticker",      ticker)
        tracker.log_param("architecture",arch_name)
        tracker.log_param("features",    len(features))
        tracker.log_metric("rmse",       rmse)
        tracker.log_metric("mae",        mae)
        tracker.log_metric("mape",       mape)
        tracker.log_metric("corr",       corr)
        tracker.end()

        results[arch_name] = {
            "model":        model,
            "scaler":       scaler,
            "preds":        preds_inv,
            "actuals":      actuals_inv,
            "train_losses": train_losses,
            "val_losses":   val_losses,
            "rmse":         rmse,
            "mae":          mae,
            "mape":         mape,
            "corr":         corr,
            "features":     features,
        }

    with open(CONFIG["models_dir"] / f"scaler_{ticker}.pkl", "wb") as f:
        pickle.dump(scaler, f)

    return results


lstm_results = {}
lstm_tickers = ["AAPL", "NVDA", "TSLA"]

for ticker in lstm_tickers:
    print("=" * 55)
    print(f"Ticker : {ticker}")
    print("=" * 55)
    lstm_results[ticker] = train_lstm_pipeline(
        ticker, epochs=80, patience=15
    )
    print()

fig = make_subplots(
    rows=len(lstm_tickers), cols=2,
    subplot_titles=[
        f"{t} — {arch}"
        for t in lstm_tickers
        for arch in ["LSTM+Attention Loss", "BiLSTM+Attention Loss"]
    ],
    vertical_spacing=0.08
)

for row, ticker in enumerate(lstm_tickers, 1):
    for col, arch in enumerate(["LSTM+Attention", "BiLSTM+Attention"], 1):
        r = lstm_results[ticker][arch]
        fig.add_trace(go.Scatter(
            y=r["train_losses"], name="Train",
            line=dict(color="#1F77B4"),
            showlegend=(row == 1 and col == 1)
        ), row=row, col=col)
        fig.add_trace(go.Scatter(
            y=r["val_losses"], name="Val",
            line=dict(color="#FF7F0E"),
            showlegend=(row == 1 and col == 1)
        ), row=row, col=col)

fig.update_layout(
    title="LSTM Training Curves — All Architectures",
    height=700, template="plotly_white",
)
fig.show()
save_fig(fig, "lstm_training_curves.png")

fig2 = make_subplots(
    rows=1, cols=len(lstm_tickers),
    subplot_titles=[f"{t} — Best Predictions" for t in lstm_tickers]
)

for col, ticker in enumerate(lstm_tickers, 1):
    best_arch = max(
        lstm_results[ticker].keys(),
        key=lambda a: lstm_results[ticker][a]["corr"]
    )
    r = lstm_results[ticker][best_arch]
    fig2.add_trace(go.Scatter(
        y=r["actuals"], name="Actual",
        line=dict(color="#1F77B4"),
        showlegend=(col == 1)
    ), row=1, col=col)
    fig2.add_trace(go.Scatter(
        y=r["preds"], name=f"Predicted ({best_arch})",
        line=dict(color="#FF7F0E", dash="dot"),
        showlegend=(col == 1)
    ), row=1, col=col)

fig2.update_layout(
    title="LSTM — Best Architecture Predictions vs Actuals",
    height=400, template="plotly_white",
)
fig2.show()
save_fig(fig2, "lstm_predictions.png")

print()
print("=" * 70)
print("LSTM Results Summary")
print("=" * 70)
print(f"{'Ticker':<8} {'Architecture':<20} {'RMSE':>10} {'MAE':>10} "
      f"{'MAPE':>10} {'Corr':>10}")
print("-" * 70)

for ticker in lstm_tickers:
    for arch, r in lstm_results[ticker].items():
        print(f"{ticker:<8} {arch:<20} {r['rmse']:>10.4f} "
              f"{r['mae']:>10.4f} {r['mape']:>9.2f}% {r['corr']:>10.4f}")
    print()

print("Cell 8 done. Tell me and I give Cell 9.")
```

```python
import optuna
import warnings
warnings.filterwarnings("ignore")
optuna.logging.set_verbosity(optuna.logging.WARNING)

from sklearn.metrics import mean_squared_error, mean_absolute_error
from sklearn.model_selection import TimeSeriesSplit
from catboost import CatBoostRegressor

def prepare_ml_features(df, horizon=5):
    feat = df.copy()

    for lag in [1, 3, 5, 10, 20]:
        feat[f"return_lag_{lag}"]  = feat["returns"].shift(lag)
        feat[f"close_lag_{lag}"]   = feat["Close"].shift(lag)
        feat[f"volume_lag_{lag}"]  = feat["Volume"].shift(lag)

    for w in [5, 10, 20, 60]:
        feat[f"rolling_mean_{w}"]  = feat["Close"].rolling(w).mean()
        feat[f"rolling_std_{w}"]   = feat["Close"].rolling(w).std()
        feat[f"rolling_min_{w}"]   = feat["Close"].rolling(w).min()
        feat[f"rolling_max_{w}"]   = feat["Close"].rolling(w).max()
        feat[f"rolling_ret_{w}"]   = feat["returns"].rolling(w).mean()
        feat[f"rolling_vol_{w}"]   = feat["returns"].rolling(w).std()

    feat["day_of_week"]  = feat.index.dayofweek
    feat["month"]        = feat.index.month
    feat["quarter"]      = feat.index.quarter
    feat["is_month_end"] = feat.index.is_month_end.astype(int)
    feat["target"]       = feat["Close"].shift(-horizon)

    base_features = [
        "Open", "High", "Low", "Close", "Volume",
        "rsi", "macd", "macd_sig", "macd_diff",
        "bb_high", "bb_low", "bb_mid", "bb_width",
        "atr", "ema_20", "ema_50", "ema_200",
        "obv", "mfi", "stoch", "williams", "roc",
        "returns", "volatility", "vol_ratio",
        "price_range", "gap", "momentum_20", "adx",
    ]
    lag_features     = [c for c in feat.columns if c.startswith(
        ("return_lag_", "close_lag_", "volume_lag_",
         "rolling_mean_", "rolling_std_", "rolling_min_",
         "rolling_max_", "rolling_ret_", "rolling_vol_")
    )]
    calendar_features = ["day_of_week", "month", "quarter", "is_month_end"]
    macro_features    = [c for c in df.columns if c.startswith("macro_")]

    feature_cols = [f for f in base_features + lag_features +
                    calendar_features + macro_features
                    if f in feat.columns]

    feat = feat.dropna()
    return feat[feature_cols], feat["target"], feature_cols


def train_xgboost(ticker, n_trials=25):
    df       = market_data[ticker]
    X, y, fc = prepare_ml_features(df, horizon=5)
    n        = len(X)
    n_train  = int(n * 0.80)
    X_train, y_train = X.iloc[:n_train], y.iloc[:n_train]
    X_test,  y_test  = X.iloc[n_train:], y.iloc[n_train:]

    def objective(trial):
        params = {
            "n_estimators":     trial.suggest_int("n_estimators", 100, 500),
            "max_depth":        trial.suggest_int("max_depth", 3, 8),
            "learning_rate":    trial.suggest_float("learning_rate", 0.01, 0.3),
            "subsample":        trial.suggest_float("subsample", 0.6, 1.0),
            "colsample_bytree": trial.suggest_float("colsample_bytree", 0.6, 1.0),
            "reg_alpha":        trial.suggest_float("reg_alpha", 0.0, 1.0),
            "reg_lambda":       trial.suggest_float("reg_lambda", 0.0, 1.0),
            "random_state":     42,
            "device":           "cuda",
            "tree_method":      "hist",
            "verbosity":        0,
        }
        tscv   = TimeSeriesSplit(n_splits=3)
        scores = []
        for tr_idx, val_idx in tscv.split(X_train):
            Xtr, Xval = X_train.iloc[tr_idx], X_train.iloc[val_idx]
            ytr, yval = y_train.iloc[tr_idx], y_train.iloc[val_idx]
            m = xgb.XGBRegressor(**params)
            m.fit(Xtr, ytr, eval_set=[(Xval, yval)], verbose=False)
            scores.append(np.sqrt(mean_squared_error(yval, m.predict(Xval))))
        return np.mean(scores)

    study = optuna.create_study(direction="minimize")
    study.optimize(objective, n_trials=n_trials, show_progress_bar=False)

    best_params = study.best_params
    best_params.update({"random_state": 42, "device": "cuda",
                        "tree_method": "hist", "verbosity": 0})
    model = xgb.XGBRegressor(**best_params)
    model.fit(X_train, y_train, verbose=False)

    preds = model.predict(X_test)
    rmse  = np.sqrt(mean_squared_error(y_test, preds))
    mae   = mean_absolute_error(y_test, preds)
    mape  = np.mean(np.abs((y_test.values - preds) / y_test.values)) * 100
    corr  = np.corrcoef(y_test.values, preds)[0, 1]

    tracker.start(f"xgboost_{ticker}")
    tracker.log_param("ticker",   ticker)
    tracker.log_metric("rmse",    rmse)
    tracker.log_metric("mae",     mae)
    tracker.log_metric("mape",    mape)
    tracker.log_metric("corr",    corr)
    tracker.end()

    return {"model": model, "preds": preds, "actuals": y_test.values,
            "rmse": rmse, "mae": mae, "mape": mape, "corr": corr,
            "feature_names": fc, "best_cv": study.best_value}


def train_lightgbm(ticker, n_trials=25):
    df       = market_data[ticker]
    X, y, fc = prepare_ml_features(df, horizon=5)
    n        = len(X)
    n_train  = int(n * 0.80)
    X_train, y_train = X.iloc[:n_train], y.iloc[:n_train]
    X_test,  y_test  = X.iloc[n_train:], y.iloc[n_train:]

    def objective(trial):
        params = {
            "n_estimators":     trial.suggest_int("n_estimators", 100, 500),
            "max_depth":        trial.suggest_int("max_depth", 3, 8),
            "learning_rate":    trial.suggest_float("learning_rate", 0.01, 0.3),
            "num_leaves":       trial.suggest_int("num_leaves", 20, 100),
            "subsample":        trial.suggest_float("subsample", 0.6, 1.0),
            "colsample_bytree": trial.suggest_float("colsample_bytree", 0.6, 1.0),
            "reg_alpha":        trial.suggest_float("reg_alpha", 0.0, 1.0),
            "reg_lambda":       trial.suggest_float("reg_lambda", 0.0, 1.0),
            "device":           "gpu",
            "verbose":          -1,
            "random_state":     42,
        }
        tscv   = TimeSeriesSplit(n_splits=3)
        scores = []
        for tr_idx, val_idx in tscv.split(X_train):
            Xtr, Xval = X_train.iloc[tr_idx], X_train.iloc[val_idx]
            ytr, yval = y_train.iloc[tr_idx], y_train.iloc[val_idx]
            m = lgb.LGBMRegressor(**params)
            m.fit(Xtr, ytr, eval_set=[(Xval, yval)])
            scores.append(np.sqrt(mean_squared_error(yval, m.predict(Xval))))
        return np.mean(scores)

    study = optuna.create_study(direction="minimize")
    study.optimize(objective, n_trials=n_trials, show_progress_bar=False)

    best_params = study.best_params
    best_params.update({"device": "gpu", "verbose": -1, "random_state": 42})
    model = lgb.LGBMRegressor(**best_params)
    model.fit(X_train, y_train)

    preds = model.predict(X_test)
    rmse  = np.sqrt(mean_squared_error(y_test, preds))
    mae   = mean_absolute_error(y_test, preds)
    mape  = np.mean(np.abs((y_test.values - preds) / y_test.values)) * 100
    corr  = np.corrcoef(y_test.values, preds)[0, 1]

    tracker.start(f"lgbm_{ticker}")
    tracker.log_param("ticker",   ticker)
    tracker.log_metric("rmse",    rmse)
    tracker.log_metric("mae",     mae)
    tracker.log_metric("mape",    mape)
    tracker.log_metric("corr",    corr)
    tracker.end()

    return {"model": model, "preds": preds, "actuals": y_test.values,
            "rmse": rmse, "mae": mae, "mape": mape, "corr": corr,
            "feature_names": fc}


def train_catboost(ticker, n_trials=20):
    df       = market_data[ticker]
    X, y, fc = prepare_ml_features(df, horizon=5)
    n        = len(X)
    n_train  = int(n * 0.80)
    X_train, y_train = X.iloc[:n_train], y.iloc[:n_train]
    X_test,  y_test  = X.iloc[n_train:], y.iloc[n_train:]

    def objective(trial):
        params = {
            "iterations":   trial.suggest_int("iterations", 100, 500),
            "depth":        trial.suggest_int("depth", 3, 8),
            "learning_rate":trial.suggest_float("learning_rate", 0.01, 0.3),
            "l2_leaf_reg":  trial.suggest_float("l2_leaf_reg", 1, 10),
            "random_seed":  42,
            "task_type":    "GPU",
            "verbose":      False,
        }
        tscv   = TimeSeriesSplit(n_splits=3)
        scores = []
        for tr_idx, val_idx in tscv.split(X_train):
            Xtr, Xval = X_train.iloc[tr_idx], X_train.iloc[val_idx]
            ytr, yval = y_train.iloc[tr_idx], y_train.iloc[val_idx]
            m = CatBoostRegressor(**params)
            m.fit(Xtr, ytr, eval_set=(Xval, yval), verbose=False)
            scores.append(np.sqrt(mean_squared_error(yval, m.predict(Xval))))
        return np.mean(scores)

    study = optuna.create_study(direction="minimize")
    study.optimize(objective, n_trials=n_trials, show_progress_bar=False)

    best_params = study.best_params
    best_params.update({"random_seed": 42, "task_type": "GPU", "verbose": False})
    model = CatBoostRegressor(**best_params)
    model.fit(X_train, y_train, verbose=False)

    preds = model.predict(X_test)
    rmse  = np.sqrt(mean_squared_error(y_test, preds))
    mae   = mean_absolute_error(y_test, preds)
    mape  = np.mean(np.abs((y_test.values - preds) / y_test.values)) * 100
    corr  = np.corrcoef(y_test.values, preds)[0, 1]

    tracker.start(f"catboost_{ticker}")
    tracker.log_param("ticker",   ticker)
    tracker.log_metric("rmse",    rmse)
    tracker.log_metric("mae",     mae)
    tracker.log_metric("mape",    mape)
    tracker.log_metric("corr",    corr)
    tracker.end()

    return {"model": model, "preds": preds, "actuals": y_test.values,
            "rmse": rmse, "mae": mae, "mape": mape, "corr": corr,
            "feature_names": fc}


xgb_results  = {}
lgbm_results = {}
cat_results  = {}
ml_tickers   = ["AAPL", "NVDA", "TSLA"]

for ticker in ml_tickers:
    print("=" * 55)
    print(f"Ticker : {ticker}")
    print("=" * 55)

    print(f"  XGBoost (25 trials)...")
    xgb_results[ticker] = train_xgboost(ticker, n_trials=25)
    print(f"    RMSE: {xgb_results[ticker]['rmse']:.4f}  "
          f"MAPE: {xgb_results[ticker]['mape']:.2f}%  "
          f"Corr: {xgb_results[ticker]['corr']:.4f}")

    print(f"  LightGBM (25 trials)...")
    lgbm_results[ticker] = train_lightgbm(ticker, n_trials=25)
    print(f"    RMSE: {lgbm_results[ticker]['rmse']:.4f}  "
          f"MAPE: {lgbm_results[ticker]['mape']:.2f}%  "
          f"Corr: {lgbm_results[ticker]['corr']:.4f}")

    print(f"  CatBoost (20 trials)...")
    cat_results[ticker] = train_catboost(ticker, n_trials=20)
    print(f"    RMSE: {cat_results[ticker]['rmse']:.4f}  "
          f"MAPE: {cat_results[ticker]['mape']:.2f}%  "
          f"Corr: {cat_results[ticker]['corr']:.4f}")
    print()

# ------------------------------------------------------------
# Feature Importance
# ------------------------------------------------------------

for ticker in ml_tickers:
    model = xgb_results[ticker]["model"]
    names = xgb_results[ticker]["feature_names"]
    imps  = model.feature_importances_
    idx   = np.argsort(imps)[::-1][:15]

    fig = go.Figure(go.Bar(
        x=imps[idx][::-1],
        y=[names[i] for i in idx][::-1],
        orientation="h",
        marker_color="#1F77B4",
    ))
    fig.update_layout(
        title=f"XGBoost Feature Importance — {ticker} (Top 15)",
        xaxis_title="Importance",
        height=500, template="plotly_white",
    )
    fig.show()
    save_fig(fig, f"feature_importance_{ticker}.png")

# ------------------------------------------------------------
# Comparaison globale
# ------------------------------------------------------------

print()
print("=" * 80)
print("Model Comparison — MAPE (%) — lower is better")
print("=" * 80)
print(f"{'Ticker':<8} {'LSTM+Att':>12} {'BiLSTM+Att':>12} "
      f"{'XGBoost':>12} {'LightGBM':>12} {'CatBoost':>12}")
print("-" * 80)

for ticker in ml_tickers:
    lstm_mape  = lstm_results[ticker]["LSTM+Attention"]["mape"]
    bilstm_mape= lstm_results[ticker]["BiLSTM+Attention"]["mape"]
    xgb_mape   = xgb_results[ticker]["mape"]
    lgbm_mape  = lgbm_results[ticker]["mape"]
    cat_mape   = cat_results[ticker]["mape"]
    print(f"{ticker:<8} {lstm_mape:>11.2f}% {bilstm_mape:>11.2f}% "
          f"{xgb_mape:>11.2f}% {lgbm_mape:>11.2f}% {cat_mape:>11.2f}%")

print()
print("=" * 80)
print("Model Comparison — Correlation — higher is better")
print("=" * 80)
print(f"{'Ticker':<8} {'LSTM+Att':>12} {'BiLSTM+Att':>12} "
      f"{'XGBoost':>12} {'LightGBM':>12} {'CatBoost':>12}")
print("-" * 80)

for ticker in ml_tickers:
    print(
        f"{ticker:<8} "
        f"{lstm_results[ticker]['LSTM+Attention']['corr']:>12.4f} "
        f"{lstm_results[ticker]['BiLSTM+Attention']['corr']:>12.4f} "
        f"{xgb_results[ticker]['corr']:>12.4f} "
        f"{lgbm_results[ticker]['corr']:>12.4f} "
        f"{cat_results[ticker]['corr']:>12.4f}"
    )

fig = make_subplots(
    rows=1, cols=2,
    subplot_titles=("MAPE (%) — lower is better",
                    "Correlation — higher is better")
)
models_names = ["LSTM+Att", "BiLSTM+Att", "XGBoost", "LightGBM", "CatBoost"]
colors_comp  = px.colors.qualitative.Plotly[:len(ml_tickers)]

for j, ticker in enumerate(ml_tickers):
    mapes = [
        lstm_results[ticker]["LSTM+Attention"]["mape"],
        lstm_results[ticker]["BiLSTM+Attention"]["mape"],
        xgb_results[ticker]["mape"],
        lgbm_results[ticker]["mape"],
        cat_results[ticker]["mape"],
    ]
    corrs = [
        lstm_results[ticker]["LSTM+Attention"]["corr"],
        lstm_results[ticker]["BiLSTM+Attention"]["corr"],
        xgb_results[ticker]["corr"],
        lgbm_results[ticker]["corr"],
        cat_results[ticker]["corr"],
    ]
    fig.add_trace(go.Bar(
        name=ticker, x=models_names, y=mapes,
        marker_color=colors_comp[j], showlegend=True,
    ), row=1, col=1)
    fig.add_trace(go.Bar(
        name=ticker, x=models_names, y=corrs,
        marker_color=colors_comp[j], showlegend=False,
    ), row=1, col=2)

fig.update_layout(
    title="FinSight AI Pro — Model Comparison (5 Architectures)",
    barmode="group", height=450, template="plotly_white",
)
fig.show()
save_fig(fig, "model_comparison.png")

print()
print("Cell 9 done. Tell me and I give Cell 10.")
```

```python
from sklearn.linear_model import Ridge
import quantstats as qs

# ------------------------------------------------------------
# PARTIE A : Ensemble model
# ------------------------------------------------------------

def build_ensemble(ticker):
    bilstm_preds = lstm_results[ticker]["BiLSTM+Attention"]["preds"]
    xgb_preds    = xgb_results[ticker]["preds"]
    lgbm_preds   = lgbm_results[ticker]["preds"]
    cat_preds    = cat_results[ticker]["preds"]
    xgb_acts     = xgb_results[ticker]["actuals"]

    n = min(len(bilstm_preds), len(xgb_preds),
            len(lgbm_preds),   len(cat_preds), len(xgb_acts))

    bilstm_preds = bilstm_preds[-n:]
    xgb_preds    = xgb_preds[-n:]
    lgbm_preds   = lgbm_preds[-n:]
    cat_preds    = cat_preds[-n:]
    actuals      = xgb_acts[-n:]

    w_bilstm = max(lstm_results[ticker]["BiLSTM+Attention"]["corr"], 0)
    w_xgb    = max(xgb_results[ticker]["corr"],  0)
    w_lgbm   = max(lgbm_results[ticker]["corr"], 0)
    w_cat    = max(cat_results[ticker]["corr"],  0)
    total    = w_bilstm + w_xgb + w_lgbm + w_cat + 1e-9

    weighted = (
        (w_bilstm / total) * bilstm_preds +
        (w_xgb    / total) * xgb_preds    +
        (w_lgbm   / total) * lgbm_preds   +
        (w_cat    / total) * cat_preds
    )

    split    = int(len(weighted) * 0.5)
    X_meta   = np.column_stack([bilstm_preds, xgb_preds, lgbm_preds, cat_preds])
    meta     = Ridge(alpha=1.0)
    meta.fit(X_meta[:split], actuals[:split])
    stacked  = meta.predict(X_meta[split:])
    acts_stk = actuals[split:]

    rmse_w = np.sqrt(mean_squared_error(actuals,  weighted))
    mape_w = np.mean(np.abs((actuals  - weighted) / actuals))  * 100
    corr_w = np.corrcoef(actuals, weighted)[0, 1]

    rmse_s = np.sqrt(mean_squared_error(acts_stk, stacked))
    mape_s = np.mean(np.abs((acts_stk - stacked)  / acts_stk)) * 100
    corr_s = np.corrcoef(acts_stk, stacked)[0, 1]

    tracker.start(f"ensemble_{ticker}")
    tracker.log_metric("weighted_rmse", rmse_w)
    tracker.log_metric("weighted_mape", mape_w)
    tracker.log_metric("weighted_corr", corr_w)
    tracker.log_metric("stacked_rmse",  rmse_s)
    tracker.log_metric("stacked_mape",  mape_s)
    tracker.log_metric("stacked_corr",  corr_s)
    tracker.end()

    return {
        "weighted":      weighted,
        "stacked":       stacked,
        "actuals":       actuals,
        "acts_stacked":  acts_stk,
        "rmse_weighted": rmse_w, "mape_weighted": mape_w, "corr_weighted": corr_w,
        "rmse_stacked":  rmse_s, "mape_stacked":  mape_s, "corr_stacked":  corr_s,
        "weights": {
            "bilstm": w_bilstm / total, "xgb": w_xgb / total,
            "lgbm":   w_lgbm   / total, "cat": w_cat / total,
        }
    }

ensemble_results = {}
print("Building ensemble models...")
print()
for ticker in ["AAPL", "NVDA", "TSLA"]:
    ensemble_results[ticker] = build_ensemble(ticker)
    r = ensemble_results[ticker]
    print(f"{ticker}")
    print(f"  Weights  : BiLSTM={r['weights']['bilstm']:.2f}  "
          f"XGB={r['weights']['xgb']:.2f}  "
          f"LGBM={r['weights']['lgbm']:.2f}  "
          f"CatBoost={r['weights']['cat']:.2f}")
    print(f"  Weighted : RMSE={r['rmse_weighted']:.4f}  "
          f"MAPE={r['mape_weighted']:.2f}%  "
          f"Corr={r['corr_weighted']:.4f}")
    print(f"  Stacked  : RMSE={r['rmse_stacked']:.4f}  "
          f"MAPE={r['mape_stacked']:.2f}%  "
          f"Corr={r['corr_stacked']:.4f}")
    print()

# ------------------------------------------------------------
# PARTIE B : Monte Carlo
# ------------------------------------------------------------

def monte_carlo_simulation(ticker, n_simulations=1000, horizon=252):
    df      = market_data[ticker]
    returns = df["returns"].dropna()
    mu      = returns.mean()
    sigma   = returns.std()
    S0      = df["Close"].iloc[-1]

    np.random.seed(42)
    simulations = np.zeros((n_simulations, horizon))
    for i in range(n_simulations):
        r      = np.random.normal(mu, sigma, horizon)
        prices = [S0]
        for ri in r:
            prices.append(prices[-1] * (1 + ri))
        simulations[i] = prices[1:]

    final_prices = simulations[:, -1]
    return {
        "simulations": simulations,
        "final_prices":final_prices,
        "S0":          S0,
        "p5":          np.percentile(final_prices, 5),
        "p25":         np.percentile(final_prices, 25),
        "p50":         np.percentile(final_prices, 50),
        "p75":         np.percentile(final_prices, 75),
        "p95":         np.percentile(final_prices, 95),
        "prob_gain":   (final_prices > S0).mean() * 100,
        "exp_return":  ((final_prices / S0) - 1).mean() * 100,
        "horizon_days":horizon,
    }

mc_results = {}
print("Running Monte Carlo simulations (1000 paths, 252 days)...")
for ticker in ["AAPL", "NVDA", "TSLA"]:
    mc_results[ticker] = monte_carlo_simulation(ticker, 1000, 252)
    r = mc_results[ticker]
    print(f"  {ticker:<6} : Current=${r['S0']:.2f}  "
          f"P50=${r['p50']:.2f}  P5=${r['p5']:.2f}  "
          f"P95=${r['p95']:.2f}  Prob(gain)={r['prob_gain']:.1f}%")

# ------------------------------------------------------------
# PARTIE C : VaR / CVaR
# ------------------------------------------------------------

print()
print("=" * 70)
print("Advanced Risk Metrics — VaR & CVaR")
print("=" * 70)
print(f"{'Ticker':<8} {'VaR 95%':>10} {'VaR 99%':>10} "
      f"{'CVaR 95%':>10} {'CVaR 99%':>10} {'MC P5 Ret':>12}")
print("-" * 70)

risk_metrics = {}
for ticker, df in market_data.items():
    returns   = df["returns"].dropna()
    var_95    = np.percentile(returns, 5)  * 100
    var_99    = np.percentile(returns, 1)  * 100
    cvar_95   = returns[returns <= np.percentile(returns, 5)].mean() * 100
    cvar_99   = returns[returns <= np.percentile(returns, 1)].mean() * 100
    mc_p5_ret = ((mc_results[ticker]["p5"] / mc_results[ticker]["S0"]) - 1) * 100 \
                if ticker in mc_results else None

    risk_metrics[ticker] = {
        "var_95": var_95, "var_99": var_99,
        "cvar_95": cvar_95, "cvar_99": cvar_99,
        "mc_p5_ret": mc_p5_ret,
    }
    mc_str = f"{mc_p5_ret:>11.1f}%" if mc_p5_ret is not None else "         N/A"
    print(f"{ticker:<8} {var_95:>9.2f}% {var_99:>9.2f}% "
          f"{cvar_95:>9.2f}% {cvar_99:>9.2f}% {mc_str}")

# ------------------------------------------------------------
# PARTIE D : Backtesting
# ------------------------------------------------------------

def backtest_strategy(ticker, model_preds, model_acts, transaction_cost=0.001):
    df         = market_data[ticker]
    close      = df["Close"].values
    n          = len(model_preds)
    close_test = close[-n - 1:-1]
    close_next = close[-n:]

    signals    = (model_preds > close_test).astype(int)
    strategy_r = np.where(signals == 1,
                          (close_next - close_test) / close_test, 0)
    strategy_r = strategy_r - transaction_cost * np.abs(
        np.diff(np.concatenate([[0], signals]))
    )
    bh_r       = (close_next - close_test) / close_test

    strat_cum  = (1 + strategy_r).cumprod()
    bh_cum     = (1 + bh_r).cumprod()

    return {
        "strat_returns": strategy_r,
        "bh_returns":    bh_r,
        "strat_cum":     strat_cum,
        "bh_cum":        bh_cum,
        "strat_total":   (strat_cum[-1] - 1) * 100,
        "bh_total":      (bh_cum[-1]   - 1) * 100,
        "strat_sharpe":  (strategy_r.mean() / (strategy_r.std() + 1e-9)) * np.sqrt(252),
        "bh_sharpe":     (bh_r.mean()       / (bh_r.std()       + 1e-9)) * np.sqrt(252),
        "strat_dd":      ((strat_cum / np.maximum.accumulate(strat_cum)) - 1).min() * 100,
        "bh_dd":         ((bh_cum   / np.maximum.accumulate(bh_cum))    - 1).min() * 100,
        "win_rate":      (strategy_r > 0).mean() * 100,
        "n_trades":      int(np.abs(np.diff(np.concatenate([[0], signals]))).sum()),
    }

backtest_results = {}
print()
print("=" * 75)
print("Backtesting Results (BiLSTM strategy vs Buy & Hold)")
print("=" * 75)
print(f"{'Ticker':<8} {'Strat Ret':>10} {'B&H Ret':>10} "
      f"{'Strat Sharpe':>14} {'B&H Sharpe':>12} "
      f"{'Strat MDD':>10} {'Win Rate':>10}")
print("-" * 75)

for ticker in ["AAPL", "NVDA", "TSLA"]:
    preds = lstm_results[ticker]["BiLSTM+Attention"]["preds"]
    acts  = lstm_results[ticker]["BiLSTM+Attention"]["actuals"]
    bt    = backtest_strategy(ticker, preds, acts)
    backtest_results[ticker] = bt
    print(f"{ticker:<8} {bt['strat_total']:>9.2f}% {bt['bh_total']:>9.2f}% "
          f"{bt['strat_sharpe']:>14.3f} {bt['bh_sharpe']:>12.3f} "
          f"{bt['strat_dd']:>9.2f}% {bt['win_rate']:>9.1f}%")

tracker.start("backtesting")
for ticker in ["AAPL", "NVDA", "TSLA"]:
    bt = backtest_results[ticker]
    tracker.log_metric(f"{ticker}_strat_ret",    bt["strat_total"])
    tracker.log_metric(f"{ticker}_bh_ret",       bt["bh_total"])
    tracker.log_metric(f"{ticker}_strat_sharpe", bt["strat_sharpe"])
    tracker.log_metric(f"{ticker}_win_rate",     bt["win_rate"])
tracker.end()

# ------------------------------------------------------------
# Visualisations
# ------------------------------------------------------------

# Fig 1 : Ensemble predictions
fig = make_subplots(
    rows=1, cols=3,
    subplot_titles=[f"{t} — Ensemble vs Actual" for t in ["AAPL", "NVDA", "TSLA"]]
)
for i, ticker in enumerate(["AAPL", "NVDA", "TSLA"], 1):
    r = ensemble_results[ticker]
    fig.add_trace(go.Scatter(
        y=r["actuals"], name="Actual",
        line=dict(color="#1F77B4", width=1.5),
        showlegend=(i == 1)
    ), row=1, col=i)
    fig.add_trace(go.Scatter(
        y=r["weighted"], name="Weighted Ensemble",
        line=dict(color="#FF7F0E", width=1.5, dash="dot"),
        showlegend=(i == 1)
    ), row=1, col=i)
fig.update_layout(
    title="Ensemble Predictions vs Actuals",
    height=400, template="plotly_white"
)
fig.show()
save_fig(fig, "ensemble_predictions.png")

# Fig 2 : Monte Carlo — lignes de percentile comme traces
fig2 = make_subplots(
    rows=1, cols=3,
    subplot_titles=[f"{t} — Monte Carlo (1000 paths)" for t in ["AAPL", "NVDA", "TSLA"]]
)
for i, ticker in enumerate(["AAPL", "NVDA", "TSLA"], 1):
    mc = mc_results[ticker]
    for j in range(0, 1000, 50):
        fig2.add_trace(go.Scatter(
            y=mc["simulations"][j],
            line=dict(color="rgba(100,100,200,0.07)", width=1),
            showlegend=False,
        ), row=1, col=i)
    for pct_key, color, label in [
        ("p5",  "red",   "P5"),
        ("p50", "black", "P50"),
        ("p95", "green", "P95"),
    ]:
        fig2.add_trace(go.Scatter(
            y=[mc[pct_key]] * mc["horizon_days"],
            name=label,
            line=dict(color=color, dash="dash", width=2),
            showlegend=(i == 1),
        ), row=1, col=i)

fig2.update_layout(
    title="Monte Carlo Price Simulation (252 trading days)",
    height=450, template="plotly_white"
)
fig2.show()
save_fig(fig2, "monte_carlo.png")

# Fig 3 : Backtest cumulative returns
fig3 = make_subplots(
    rows=1, cols=3,
    subplot_titles=[f"{t} — Strategy vs Buy & Hold" for t in ["AAPL", "NVDA", "TSLA"]]
)
for i, ticker in enumerate(["AAPL", "NVDA", "TSLA"], 1):
    bt = backtest_results[ticker]
    fig3.add_trace(go.Scatter(
        y=bt["strat_cum"], name="Strategy",
        line=dict(color="#26A69A", width=2),
        showlegend=(i == 1)
    ), row=1, col=i)
    fig3.add_trace(go.Scatter(
        y=bt["bh_cum"], name="Buy & Hold",
        line=dict(color="#EF5350", width=2, dash="dot"),
        showlegend=(i == 1)
    ), row=1, col=i)
fig3.update_layout(
    title="Strategy vs Buy & Hold (BiLSTM signals)",
    height=400, template="plotly_white"
)
fig3.show()
save_fig(fig3, "backtest_comparison.png")

# Fig 4 : Monte Carlo distribution
fig4 = go.Figure()
for ticker in ["AAPL", "NVDA", "TSLA"]:
    mc = mc_results[ticker]
    fig4.add_trace(go.Histogram(
        x=((mc["final_prices"] / mc["S0"]) - 1) * 100,
        name=ticker, opacity=0.6, nbinsx=50,
    ))
fig4.update_layout(
    title="Monte Carlo Final Return Distribution (252 days)",
    xaxis_title="Return (%)", yaxis_title="Frequency",
    barmode="overlay", height=400, template="plotly_white"
)
fig4.show()
save_fig(fig4, "monte_carlo_distribution.png")

print()
print("=" * 70)
print("Cell 10 Summary")
print("=" * 70)
print("Ensemble      : Weighted average + Ridge stacking")
print("Monte Carlo   : 1000 simulations x 252 days")
print()
for ticker in ["AAPL", "NVDA", "TSLA"]:
    r  = ensemble_results[ticker]
    mc = mc_results[ticker]
    bt = backtest_results[ticker]
    print(f"{ticker}")
    print(f"  Ensemble MAPE   : {r['mape_weighted']:.2f}%  "
          f"Corr={r['corr_weighted']:.4f}")
    print(f"  MC Expected Ret : {mc['exp_return']:.1f}%  "
          f"P5=${mc['p5']:.2f}  P95=${mc['p95']:.2f}")
    print(f"  Strategy Ret    : {bt['strat_total']:.2f}%  "
          f"vs B&H {bt['bh_total']:.2f}%  "
          f"Sharpe={bt['strat_sharpe']:.3f}")
    print()

print("Cell 10 done. Tell me and I give Cell 11.")
```

```python
# Cell 11 — Market Regime Detection avancee (HMM + GMM + Visualisations)

from hmmlearn import hmm
from sklearn.mixture import GaussianMixture
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings("ignore")

def detect_regimes_hmm(ticker, n_regimes=3, n_iter=200):
    df       = market_data[ticker]
    features = df[["returns", "vol_20", "rsi", "macd",
                   "adx", "obv", "mfi"]].dropna()

    scaler  = StandardScaler()
    X       = scaler.fit_transform(features)

    model   = hmm.GaussianHMM(
        n_components=n_regimes,
        covariance_type="full",
        n_iter=n_iter,
        random_state=42,
    )
    model.fit(X)
    states  = model.predict(X)
    proba   = model.predict_proba(X)

    state_returns = {}
    state_vols    = {}
    for s in range(n_regimes):
        mask              = states == s
        state_returns[s]  = features["returns"].values[mask].mean() * 252 * 100
        state_vols[s]     = features["returns"].values[mask].std()  * np.sqrt(252) * 100

    sorted_states  = sorted(state_returns.items(), key=lambda x: x[1])
    regime_map     = {
        sorted_states[0][0]: "BEAR",
        sorted_states[1][0]: "SIDEWAYS",
        sorted_states[2][0]: "BULL",
    }

    regime_colors  = {"BULL": "#26A69A", "BEAR": "#EF5350", "SIDEWAYS": "#FFA726"}
    regime_labels  = [regime_map[s] for s in states]
    current_regime = regime_map[states[-1]]
    current_proba  = {regime_map[s]: round(proba[-1][s] * 100, 1)
                      for s in range(n_regimes)}

    transition_matrix = model.transmat_
    regime_stats      = {}
    for s in range(n_regimes):
        name = regime_map[s]
        mask = states == s
        regime_stats[name] = {
            "n_days":      int(mask.sum()),
            "pct_time":    round(mask.mean() * 100, 1),
            "ann_return":  round(state_returns[s], 2),
            "ann_vol":     round(state_vols[s],    2),
            "sharpe":      round(state_returns[s] / (state_vols[s] + 1e-9), 3),
        }

    return {
        "states":          states,
        "proba":           proba,
        "regime_labels":   regime_labels,
        "regime_map":      regime_map,
        "regime_stats":    regime_stats,
        "regime_colors":   regime_colors,
        "current_regime":  current_regime,
        "current_proba":   current_proba,
        "transition":      transition_matrix,
        "dates":           features.index,
        "returns":         features["returns"].values,
        "close":           df["Close"].loc[features.index].values,
        "n_regimes":       n_regimes,
    }


def detect_regimes_gmm(ticker, n_regimes=3):
    df       = market_data[ticker]
    features = df[["returns", "vol_20", "rsi", "macd"]].dropna()

    scaler   = StandardScaler()
    X        = scaler.fit_transform(features)

    gmm      = GaussianMixture(
        n_components=n_regimes,
        covariance_type="full",
        random_state=42,
        n_init=5,
    )
    gmm.fit(X)
    states   = gmm.predict(X)
    proba    = gmm.predict_proba(X)

    state_returns = {}
    for s in range(n_regimes):
        mask             = states == s
        state_returns[s] = features["returns"].values[mask].mean() * 252 * 100

    sorted_s    = sorted(state_returns.items(), key=lambda x: x[1])
    regime_map  = {
        sorted_s[0][0]: "BEAR",
        sorted_s[1][0]: "SIDEWAYS",
        sorted_s[2][0]: "BULL",
    }
    regime_labels  = [regime_map[s] for s in states]
    current_regime = regime_map[states[-1]]

    agreement = np.mean(
        np.array(regime_labels) ==
        np.array(hmm_results[ticker]["regime_labels"][-len(regime_labels):])
    ) * 100

    return {
        "states":         states,
        "proba":          proba,
        "regime_labels":  regime_labels,
        "regime_map":     regime_map,
        "current_regime": current_regime,
        "agreement_hmm":  round(agreement, 1),
        "dates":          features.index,
    }


hmm_results = {}
gmm_results = {}

print("=" * 65)
print("Market Regime Detection — HMM + GMM")
print("=" * 65)
print()

for ticker in CONFIG["market_tickers"]:
    print(f"Analyzing {ticker}...")
    hmm_results[ticker] = detect_regimes_hmm(ticker, n_regimes=3)
    gmm_results[ticker] = detect_regimes_gmm(ticker, n_regimes=3)

    h = hmm_results[ticker]
    g = gmm_results[ticker]

    print(f"  HMM Current regime   : {h['current_regime']}")
    print(f"  HMM Probabilities    : {h['current_proba']}")
    print(f"  GMM Current regime   : {g['current_regime']}")
    print(f"  HMM-GMM Agreement    : {g['agreement_hmm']:.1f}%")
    print()
    for regime, stats in h["regime_stats"].items():
        print(f"  {regime:<10} : {stats['n_days']:>4} days "
              f"({stats['pct_time']:>5.1f}%) | "
              f"AnnRet={stats['ann_return']:>7.2f}% | "
              f"AnnVol={stats['ann_vol']:>6.2f}% | "
              f"Sharpe={stats['sharpe']:>6.3f}")
    print()

# ------------------------------------------------------------
# Visualisation 1 : Regime timeline pour AAPL NVDA TSLA
# ------------------------------------------------------------

viz_tickers = ["AAPL", "NVDA", "TSLA"]
fig = make_subplots(
    rows=len(viz_tickers), cols=1,
    subplot_titles=[f"{t} — HMM Market Regimes" for t in viz_tickers],
    vertical_spacing=0.08,
    shared_xaxes=False,
)

for row, ticker in enumerate(viz_tickers, 1):
    h      = hmm_results[ticker]
    dates  = h["dates"]
    close  = h["close"]
    labels = h["regime_labels"]
    colors = h["regime_colors"]

    fig.add_trace(go.Scatter(
        x=dates, y=close,
        name=ticker,
        line=dict(color="#333333", width=1),
        showlegend=False,
    ), row=row, col=1)

    regime_change = [0] + [i for i in range(1, len(labels))
                           if labels[i] != labels[i-1]] + [len(labels)]

    for j in range(len(regime_change) - 1):
        start = regime_change[j]
        end   = regime_change[j + 1]
        reg   = labels[start]
        fig.add_vrect(
            x0=dates[start],
            x1=dates[min(end, len(dates)-1)],
            fillcolor=colors[reg],
            opacity=0.15,
            layer="below",
            line_width=0,
            row=row, col=1,
        )

fig.update_layout(
    title="HMM Market Regime Detection (BULL / SIDEWAYS / BEAR)",
    height=750, template="plotly_white",
)
fig.show()
save_fig(fig, "regime_timeline.png")

# ------------------------------------------------------------
# Visualisation 2 : Probabilites de regime (AAPL)
# ------------------------------------------------------------

ticker = "AAPL"
h      = hmm_results[ticker]
dates  = h["dates"]
proba  = h["proba"]
colors_r = ["#26A69A", "#FFA726", "#EF5350"]
labels_r = ["BULL", "SIDEWAYS", "BEAR"]
sorted_s = sorted(h["regime_map"].items(), key=lambda x: x[1])
order    = [s for s, _ in sorted(h["regime_map"].items(),
            key=lambda x: ["BULL", "SIDEWAYS", "BEAR"].index(x[1]))]

fig2 = go.Figure()
for i, (state_idx, color, label) in enumerate(
        zip(order, colors_r, labels_r)):
    fig2.add_trace(go.Scatter(
        x=dates,
        y=proba[:, state_idx],
        name=label,
        stackgroup="one",
        line=dict(width=0),
        fillcolor=color.replace(")", ", 0.6)").replace("rgb", "rgba")
                  if color.startswith("rgb") else color,
        mode="lines",
    ))
fig2.update_layout(
    title=f"HMM Regime Probabilities — {ticker}",
    yaxis_title="Probability",
    height=400, template="plotly_white",
)
fig2.show()
save_fig(fig2, "regime_probabilities.png")

# ------------------------------------------------------------
# Visualisation 3 : Matrice de transition (AAPL)
# ------------------------------------------------------------

trans  = hmm_results[ticker]["transition"]
regime_order = ["BULL", "SIDEWAYS", "BEAR"]
regime_idx   = [k for name in regime_order
                for k, v in hmm_results[ticker]["regime_map"].items() if v == name]

trans_ordered = trans[np.ix_(regime_idx, regime_idx)]

fig3 = go.Figure(go.Heatmap(
    z=trans_ordered,
    x=regime_order,
    y=regime_order,
    colorscale="Blues",
    text=[[f"{v:.3f}" for v in row] for row in trans_ordered],
    texttemplate="%{text}",
    textfont=dict(size=14),
    zmin=0, zmax=1,
))
fig3.update_layout(
    title=f"HMM Transition Matrix — {ticker}",
    xaxis_title="To Regime",
    yaxis_title="From Regime",
    height=400, template="plotly_white",
)
fig3.show()
save_fig(fig3, "regime_transition_matrix.png")

# ------------------------------------------------------------
# Visualisation 4 : Comparaison regimes tous tickers
# ------------------------------------------------------------

regime_summary = {}
for ticker in CONFIG["market_tickers"]:
    h = hmm_results[ticker]
    regime_summary[ticker] = h["current_regime"]

fig4 = go.Figure()
regime_colors_bar = {
    "BULL":     "#26A69A",
    "SIDEWAYS": "#FFA726",
    "BEAR":     "#EF5350",
}
for regime in ["BULL", "SIDEWAYS", "BEAR"]:
    tickers_r = [t for t, r in regime_summary.items() if r == regime]
    if tickers_r:
        fig4.add_trace(go.Bar(
            x=tickers_r,
            y=[1] * len(tickers_r),
            name=regime,
            marker_color=regime_colors_bar[regime],
        ))
fig4.update_layout(
    title="Current Market Regime — All Tickers",
    yaxis=dict(showticklabels=False),
    barmode="stack",
    height=350, template="plotly_white",
)
fig4.show()
save_fig(fig4, "regime_all_tickers.png")

# ------------------------------------------------------------
# Summary
# ------------------------------------------------------------

tracker.start("regime_detection")
for ticker in CONFIG["market_tickers"]:
    h = hmm_results[ticker]
    tracker.log_param(f"{ticker}_regime",  h["current_regime"])
    for reg, stats in h["regime_stats"].items():
        tracker.log_metric(f"{ticker}_{reg}_days", stats["n_days"])
tracker.end()

print("=" * 65)
print("Cell 11 Summary — Regime Detection")
print("=" * 65)
print(f"{'Ticker':<8} {'HMM Regime':<12} {'GMM Regime':<12} {'Agreement':>10}")
print("-" * 50)
for ticker in CONFIG["market_tickers"]:
    h = hmm_results[ticker]
    g = gmm_results[ticker]
    print(f"{ticker:<8} {h['current_regime']:<12} "
          f"{g['current_regime']:<12} {g['agreement_hmm']:>9.1f}%")

print()
print("Cell 11 done. Tell me and I give Cell 12.")
```

```python
# Cell 12 — FinBERT Sentiment Analysis avance + Fine-tuning simulé

import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import json
from pathlib import Path

# ------------------------------------------------------------
# PARTIE A : FinBERT-style Sentiment Classifier custom
# (sans sentence-transformers ni transformers pipeline)
# ------------------------------------------------------------

class FinancialSentimentClassifier(nn.Module):
    """
    Classifieur de sentiment financier base sur notre embedder custom.
    Architecture : Embeddings -> BiGRU -> Attention -> MLP -> 3 classes
    """
    def __init__(self, embed_dim=384, hidden_dim=128, n_classes=3, dropout=0.3):
        super().__init__()
        self.bigru = nn.GRU(
            input_size=embed_dim,
            hidden_size=hidden_dim,
            num_layers=2,
            batch_first=True,
            bidirectional=True,
            dropout=dropout,
        )
        self.attention = nn.Sequential(
            nn.Linear(hidden_dim * 2, hidden_dim),
            nn.Tanh(),
            nn.Linear(hidden_dim, 1),
            nn.Softmax(dim=1),
        )
        self.classifier = nn.Sequential(
            nn.Linear(hidden_dim * 2, hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, n_classes),
        )

    def forward(self, x):
        # x : (batch, seq_len, embed_dim)
        out, _       = self.bigru(x)
        attn_weights = self.attention(out)
        context      = (attn_weights * out).sum(dim=1)
        return self.classifier(context)


# ------------------------------------------------------------
# PARTIE B : Dataset financier pour le fine-tuning
# ------------------------------------------------------------

FINANCIAL_TRAINING_DATA = [
    # Positif
    ("Strong earnings beat with record revenue growth",           "positive"),
    ("Stock surges on better than expected quarterly results",     "positive"),
    ("Company raises full year guidance above analyst estimates",  "positive"),
    ("Dividend increase announced boosting shareholder returns",   "positive"),
    ("Acquisition expected to be immediately accretive to EPS",   "positive"),
    ("Breakthrough product launch drives strong market demand",    "positive"),
    ("Management buyback program signals confidence in stock",     "positive"),
    ("Revenue growth accelerates driven by new market expansion",  "positive"),
    ("Analyst upgrades stock to buy with significant upside",      "positive"),
    ("Operating margins improve sharply on cost efficiency gains", "positive"),
    ("Free cash flow generation at record levels this quarter",    "positive"),
    ("Market share gains accelerate in key growth segments",       "positive"),
    ("Partnership deal unlocks substantial addressable market",    "positive"),
    ("CEO confident in long term growth trajectory ahead",         "positive"),
    ("Return on equity reaches multi year high on strong profits", "positive"),

    # Negatif
    ("Company misses earnings estimates on weak demand outlook",   "negative"),
    ("Stock falls sharply after guidance cut below expectations",  "negative"),
    ("Regulatory investigation threatens core business model",     "negative"),
    ("Supply chain disruptions weigh heavily on profit margins",   "negative"),
    ("Management warns of significant headwinds in next quarter",  "negative"),
    ("Layoffs announced as company struggles with rising costs",   "negative"),
    ("Credit rating downgraded amid deteriorating balance sheet",  "negative"),
    ("Product recall impacts revenue and damages brand reputation","negative"),
    ("Lawsuit settlement will materially impact quarterly results","negative"),
    ("Rising interest rates compress margins across all segments", "negative"),
    ("Competition intensifies putting pressure on pricing power",  "negative"),
    ("Inventory build suggests demand weakness ahead",             "negative"),
    ("Debt levels rising unsustainably fast this fiscal year",     "negative"),
    ("Operating cash flow turns negative on high capital spending","negative"),
    ("Analyst downgrades stock citing valuation concerns",         "negative"),

    # Neutre
    ("Company reports results in line with analyst expectations",  "neutral"),
    ("Management reaffirms full year guidance unchanged",          "neutral"),
    ("Board of directors schedules annual shareholder meeting",    "neutral"),
    ("Company files quarterly report with securities regulator",   "neutral"),
    ("CFO discusses capital allocation strategy in investor call", "neutral"),
    ("Merger integration proceeding according to original plan",   "neutral"),
    ("Research and development spending maintained at prior levels","neutral"),
    ("Company announces leadership transition at senior level",    "neutral"),
    ("Market conditions remain stable for core business segments", "neutral"),
    ("Quarterly results show mixed performance across divisions",  "neutral"),
    ("Stock trades sideways as investors await more clarity",      "neutral"),
    ("Volume in line with thirty day average on no major news",    "neutral"),
    ("Company explores strategic alternatives for non-core assets","neutral"),
    ("Industry headwinds balanced by internal efficiency gains",   "neutral"),
    ("Management provides update on ongoing restructuring plan",   "neutral"),
]

label_map  = {"negative": 0, "neutral": 1, "positive": 2}
label_names= ["negative", "neutral", "positive"]
colors_sent= {"positive": "#26A69A", "negative": "#EF5350", "neutral": "#FFA726"}

print("Embedding training data...")
train_texts  = [t for t, _ in FINANCIAL_TRAINING_DATA]
train_labels = [label_map[l] for _, l in FINANCIAL_TRAINING_DATA]

train_embeddings = embedder.embed(train_texts)
print(f"Training embeddings shape : {train_embeddings.shape}")

# ------------------------------------------------------------
# PARTIE C : Fine-tuning du classifieur
# ------------------------------------------------------------

from torch.utils.data import TensorDataset, DataLoader

X_train = torch.FloatTensor(train_embeddings).unsqueeze(1)
y_train = torch.LongTensor(train_labels)

dataset    = TensorDataset(X_train, y_train)
dataloader = DataLoader(dataset, batch_size=8, shuffle=True)

sentiment_model = FinancialSentimentClassifier(
    embed_dim=384, hidden_dim=128, n_classes=3, dropout=0.3
).to(CONFIG["device"])

optimizer  = torch.optim.Adam(sentiment_model.parameters(), lr=1e-3)
scheduler  = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=50)
criterion  = nn.CrossEntropyLoss()

print("\nFine-tuning FinancialSentimentClassifier...")
best_acc    = 0
train_losses= []

for epoch in range(100):
    sentiment_model.train()
    total_loss  = 0
    correct     = 0
    total       = 0

    for xb, yb in dataloader:
        xb, yb = xb.to(CONFIG["device"]), yb.to(CONFIG["device"])
        optimizer.zero_grad()
        logits = sentiment_model(xb)
        loss   = criterion(logits, yb)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()
        preds      = logits.argmax(dim=1)
        correct    += (preds == yb).sum().item()
        total      += len(yb)

    scheduler.step()
    acc = correct / total * 100
    train_losses.append(total_loss / len(dataloader))

    if acc > best_acc:
        best_acc = acc
        torch.save(
            sentiment_model.state_dict(),
            CONFIG["models_dir"] / "finbert_custom.pt"
        )

    if (epoch + 1) % 25 == 0:
        print(f"  Epoch {epoch+1:3d} | Loss: {total_loss/len(dataloader):.4f} "
              f"| Acc: {acc:.1f}%")

print(f"\nBest training accuracy : {best_acc:.1f}%")
sentiment_model.load_state_dict(
    torch.load(CONFIG["models_dir"] / "finbert_custom.pt")
)

# ------------------------------------------------------------
# PARTIE D : Inference sur les news
# ------------------------------------------------------------

def predict_sentiment(text):
    emb    = embedder.embed([text])
    x      = torch.FloatTensor(emb).unsqueeze(1).to(CONFIG["device"])
    sentiment_model.eval()
    with torch.no_grad():
        logits = sentiment_model(x)
        proba  = F.softmax(logits, dim=1).cpu().numpy()[0]
        pred   = proba.argmax()
    return {
        "label":    label_names[pred],
        "score":    float(proba[pred]),
        "proba":    {l: round(float(p), 3)
                     for l, p in zip(label_names, proba)},
    }


# Analyse des headlines news existantes
print("\nRunning FinBERT-style inference on financial news...")
print("=" * 70)

finbert_results = []
sample_news     = news_df.head(30) if not news_df.empty else pd.DataFrame()

if not sample_news.empty:
    for _, row in sample_news.iterrows():
        text   = str(row.get("title", ""))[:200]
        if len(text) < 10:
            continue
        result = predict_sentiment(text)
        result["text"]   = text
        result["source"] = row.get("source", "")
        result["ticker"] = (row["tickers_mentioned"][0]
                            if isinstance(row.get("tickers_mentioned"), list)
                            and len(row["tickers_mentioned"]) > 0
                            else "MARKET")
        finbert_results.append(result)

for r in finbert_results[:10]:
    icon = "🟢" if r["label"] == "positive" \
        else "🔴" if r["label"] == "negative" else "🟡"
    print(f"{icon} [{r['label'].upper():<8}] ({r['score']:.2f}) "
          f"{r['text'][:65]}")

# Ticker-level aggregation
print("\nFinBERT Sentiment by Ticker :")
print("=" * 60)

ticker_finbert = {}
for ticker in CONFIG["market_tickers"]:
    ticker_news = [r for r in finbert_results if r["ticker"] == ticker]
    if not ticker_news:
        ticker_news = finbert_results

    avg_pos  = np.mean([r["proba"]["positive"] for r in ticker_news])
    avg_neg  = np.mean([r["proba"]["negative"] for r in ticker_news])
    avg_neu  = np.mean([r["proba"]["neutral"]  for r in ticker_news])
    compound = avg_pos - avg_neg
    label    = "positive" if compound > 0.1 \
        else "negative" if compound < -0.1 else "neutral"

    ticker_finbert[ticker] = {
        "positive": round(avg_pos,  3),
        "negative": round(avg_neg,  3),
        "neutral":  round(avg_neu,  3),
        "compound": round(compound, 3),
        "label":    label,
    }
    print(f"  {ticker:<6} : compound={compound:>+.3f}  "
          f"pos={avg_pos:.3f}  neg={avg_neg:.3f}  {label.upper()}")

with open(CONFIG["results_dir"] / "finbert_sentiment.json", "w") as f:
    json.dump(ticker_finbert, f, indent=2)

# ------------------------------------------------------------
# Visualisations
# ------------------------------------------------------------

# Fig 1 : Training loss
fig = go.Figure()
fig.add_trace(go.Scatter(
    y=train_losses, mode="lines",
    line=dict(color="#1F77B4", width=2),
    name="Training Loss"
))
fig.update_layout(
    title="FinBERT Custom — Training Loss Curve",
    xaxis_title="Epoch", yaxis_title="Loss",
    height=350, template="plotly_white"
)
fig.show()
save_fig(fig, "finbert_training_loss.png")

# Fig 2 : Sentiment distribution
if finbert_results:
    label_counts = {"positive": 0, "negative": 0, "neutral": 0}
    for r in finbert_results:
        label_counts[r["label"]] += 1

    fig2 = go.Figure(go.Pie(
        labels=list(label_counts.keys()),
        values=list(label_counts.values()),
        marker_colors=[colors_sent[l] for l in label_counts.keys()],
        textinfo="label+percent",
    ))
    fig2.update_layout(
        title="FinBERT — News Sentiment Distribution",
        height=400, template="plotly_white"
    )
    fig2.show()
    save_fig(fig2, "finbert_distribution.png")

# Fig 3 : Sentiment par ticker (stacked bar)
tickers_sorted = sorted(
    ticker_finbert.keys(),
    key=lambda t: ticker_finbert[t]["compound"], reverse=True
)
fig3 = go.Figure()
for label, color in [
    ("positive", "#26A69A"),
    ("neutral",  "#FFA726"),
    ("negative", "#EF5350"),
]:
    fig3.add_trace(go.Bar(
        name=label.capitalize(),
        x=tickers_sorted,
        y=[ticker_finbert[t][label] for t in tickers_sorted],
        marker_color=color,
    ))
fig3.update_layout(
    title="FinBERT Sentiment Score by Ticker",
    barmode="stack",
    yaxis_title="Probability",
    height=400, template="plotly_white",
)
fig3.show()
save_fig(fig3, "finbert_by_ticker.png")

# Fig 4 : Compound score
fig4 = go.Figure(go.Bar(
    x=tickers_sorted,
    y=[ticker_finbert[t]["compound"] for t in tickers_sorted],
    marker_color=[
        colors_sent[ticker_finbert[t]["label"]] for t in tickers_sorted
    ],
    text=[f"{ticker_finbert[t]['compound']:+.3f}" for t in tickers_sorted],
    textposition="outside",
))
fig4.update_layout(
    title="FinBERT Compound Sentiment Score by Ticker",
    yaxis_title="Compound Score",
    height=400, template="plotly_white",
)
fig4.show()
save_fig(fig4, "finbert_compound.png")

tracker.start("finbert_sentiment")
tracker.log_metric("best_train_acc",  best_acc)
tracker.log_metric("n_training_samples", len(FINANCIAL_TRAINING_DATA))
tracker.log_param("architecture",    "BiGRU+Attention")
tracker.log_param("embed_dim",       384)
for ticker, s in ticker_finbert.items():
    tracker.log_metric(f"{ticker}_compound", s["compound"])
tracker.end()

print()
print("=" * 65)
print("Cell 12 Summary — FinBERT Custom Sentiment")
print("=" * 65)
print(f"Architecture        : Embedder + BiGRU + Attention + MLP")
print(f"Training samples    : {len(FINANCIAL_TRAINING_DATA)}")
print(f"Best train accuracy : {best_acc:.1f}%")
print(f"News analyzed       : {len(finbert_results)}")
print()
print("Most positive tickers :")
top3 = sorted(ticker_finbert.items(),
              key=lambda x: x[1]["compound"], reverse=True)[:3]
for t, s in top3:
    print(f"  {t:<6} : {s['compound']:+.3f}  ({s['label']})")
print()
print("Most negative tickers :")
bot3 = sorted(ticker_finbert.items(), key=lambda x: x[1]["compound"])[:3]
for t, s in bot3:
    print(f"  {t:<6} : {s['compound']:+.3f}  ({s['label']})")
print()
print("Cell 12 done. Tell me and I give Cell 13.")
```

```python
# Cell 13 — Final Report + Dashboard + Gradio Interface

import gradio as gr
import json
import time
import numpy as np

# ------------------------------------------------------------
# PARTIE A : Rapport final consolide
# ------------------------------------------------------------

def generate_final_report():
    report = {
        "project":      "FinSight AI Pro",
        "date":         datetime.now().strftime("%Y-%m-%d %H:%M"),
        "tickers":      CONFIG["market_tickers"],
        "period":       f"{CONFIG['start_date']} -> {CONFIG['end_date']}",
        "models":       {},
        "regimes":      {},
        "sentiment":    {},
        "risk":         {},
        "recommendations": {},
    }

    for ticker in ["AAPL", "NVDA", "TSLA"]:
        report["models"][ticker] = {
            "bilstm_mape":    lstm_results[ticker]["BiLSTM+Attention"]["mape"],
            "bilstm_corr":    lstm_results[ticker]["BiLSTM+Attention"]["corr"],
            "ensemble_mape":  ensemble_results[ticker]["mape_weighted"],
            "ensemble_corr":  ensemble_results[ticker]["corr_weighted"],
            "stacked_mape":   ensemble_results[ticker]["mape_stacked"],
            "stacked_corr":   ensemble_results[ticker]["corr_stacked"],
            "mc_p50":         mc_results[ticker]["p50"],
            "mc_p5":          mc_results[ticker]["p5"],
            "mc_p95":         mc_results[ticker]["p95"],
            "prob_gain":      mc_results[ticker]["prob_gain"],
        }

    for ticker in CONFIG["market_tickers"]:
        report["regimes"][ticker] = {
            "hmm": hmm_results[ticker]["current_regime"],
            "gmm": gmm_results[ticker]["current_regime"],
        }
        report["sentiment"][ticker] = ticker_finbert.get(ticker, {})
        report["risk"][ticker]      = risk_metrics.get(ticker, {})

    for ticker in ["AAPL", "NVDA", "TSLA"]:
        if ticker in agent_results:
            report["recommendations"][ticker] = {
                "signal":     agent_results[ticker]["recommendation"],
                "confidence": agent_results[ticker]["confidence"],
                "regime":     agent_results[ticker]["regime"],
            }

    with open(CONFIG["results_dir"] / "final_report.json", "w") as f:
        json.dump(report, f, indent=2, default=str)

    return report

final_report = generate_final_report()
print("Final report generated.")

# ------------------------------------------------------------
# PARTIE B : Dashboard interactif Plotly
# ------------------------------------------------------------

from plotly.subplots import make_subplots

print("Building final dashboard...")

fig_dash = make_subplots(
    rows=3, cols=3,
    subplot_titles=[
        "Price Performance (Normalized)",
        "Sharpe Ratio by Ticker",
        "VaR 95% by Ticker",
        "BiLSTM MAPE by Ticker",
        "Ensemble Correlation",
        "Monte Carlo — Prob of Gain",
        "HMM Regime Distribution",
        "FinBERT Sentiment",
        "Agent Recommendations",
    ],
    vertical_spacing=0.12,
    horizontal_spacing=0.08,
)

colors_tickers = px.colors.qualitative.Plotly

# Panel 1 : Prix normalises
for i, (ticker, df) in enumerate(market_data.items()):
    norm = df["Close"].tail(252) / df["Close"].tail(252).iloc[0] * 100
    fig_dash.add_trace(go.Scatter(
        x=list(range(len(norm))), y=norm.values,
        name=ticker,
        line=dict(color=colors_tickers[i % 10], width=1.2),
        showlegend=False,
    ), row=1, col=1)

# Panel 2 : Sharpe ratio
sharpe_vals = []
for ticker, df in market_data.items():
    r = df["returns"].dropna()
    sharpe_vals.append((returns.mean() * 252) / (r.std() * np.sqrt(252))
                       if (s := r.std() * np.sqrt(252)) > 0 else 0)

sharpe_data = [(t, (market_data[t]["returns"].dropna().mean() * 252) /
                   (market_data[t]["returns"].dropna().std() * np.sqrt(252)))
               for t in CONFIG["market_tickers"]]
sharpe_data.sort(key=lambda x: x[1], reverse=True)

fig_dash.add_trace(go.Bar(
    x=[t for t, _ in sharpe_data],
    y=[s for _, s in sharpe_data],
    marker_color="#1F77B4",
    showlegend=False,
), row=1, col=2)

# Panel 3 : VaR 95%
fig_dash.add_trace(go.Bar(
    x=list(risk_metrics.keys()),
    y=[risk_metrics[t]["var_95"] for t in risk_metrics],
    marker_color="#EF5350",
    showlegend=False,
), row=1, col=3)

# Panel 4 : BiLSTM MAPE
mape_tickers = ["AAPL", "NVDA", "TSLA"]
mape_vals    = [lstm_results[t]["BiLSTM+Attention"]["mape"] for t in mape_tickers]
fig_dash.add_trace(go.Bar(
    x=mape_tickers, y=mape_vals,
    marker_color="#FF7F0E",
    showlegend=False,
), row=2, col=1)

# Panel 5 : Ensemble correlation
corr_vals = [ensemble_results[t]["corr_weighted"] for t in mape_tickers]
fig_dash.add_trace(go.Bar(
    x=mape_tickers, y=corr_vals,
    marker_color="#2CA02C",
    showlegend=False,
), row=2, col=2)

# Panel 6 : Monte Carlo prob gain
mc_tickers = ["AAPL", "NVDA", "TSLA"]
prob_gains  = [mc_results[t]["prob_gain"] for t in mc_tickers]
fig_dash.add_trace(go.Bar(
    x=mc_tickers, y=prob_gains,
    marker_color="#9467BD",
    showlegend=False,
), row=2, col=3)

# Panel 7 : HMM regimes
regime_counts = {"BULL": 0, "SIDEWAYS": 0, "BEAR": 0}
for ticker in CONFIG["market_tickers"]:
    reg = hmm_results[ticker]["current_regime"]
    regime_counts[reg] = regime_counts.get(reg, 0) + 1

fig_dash.add_trace(go.Bar(
    x=list(regime_counts.keys()),
    y=list(regime_counts.values()),
    marker_color=["#26A69A", "#FFA726", "#EF5350"],
    showlegend=False,
), row=3, col=1)

# Panel 8 : FinBERT sentiment
sent_tickers = list(ticker_finbert.keys())
sent_vals    = [ticker_finbert[t]["compound"] for t in sent_tickers]
sent_colors  = ["#26A69A" if v > 0.1 else "#EF5350" if v < -0.1
                else "#FFA726" for v in sent_vals]
fig_dash.add_trace(go.Bar(
    x=sent_tickers, y=sent_vals,
    marker_color=sent_colors,
    showlegend=False,
), row=3, col=2)

# Panel 9 : Agent recommendations
rec_map_num = {
    "STRONG BUY": 2, "BUY": 1, "HOLD": 0, "SELL": -1, "STRONG SELL": -2
}
for ticker in ["AAPL", "NVDA", "TSLA"]:
    if ticker in agent_results:
        rec  = agent_results[ticker]["recommendation"]
        val  = rec_map_num.get(rec, 0)
        col  = "#26A69A" if val > 0 else "#EF5350" if val < 0 else "#FFA726"
        fig_dash.add_trace(go.Bar(
            x=[ticker], y=[val],
            marker_color=col,
            text=[rec], textposition="outside",
            showlegend=False,
        ), row=3, col=3)

fig_dash.update_layout(
    title="FinSight AI Pro — Final Dashboard",
    height=1000, template="plotly_white",
)
fig_dash.show()
save_fig(fig_dash, "final_dashboard.png")
print("Dashboard saved.")

# ------------------------------------------------------------
# PARTIE C : Interface Gradio
# ------------------------------------------------------------

def analyze_ticker_gradio(ticker, question):
    ticker = ticker.upper().strip()

    if ticker not in market_data:
        return f"Ticker {ticker} not found. Available: {CONFIG['market_tickers']}"

    df      = market_data[ticker]
    returns = df["returns"].dropna()
    last    = df.iloc[-1]

    ann_ret = returns.mean() * 252 * 100
    ann_vol = returns.std() * np.sqrt(252) * 100
    sharpe  = ann_ret / (ann_vol + 1e-9)
    max_dd  = ((df["Close"] / df["Close"].cummax()) - 1).min() * 100
    var_95  = np.percentile(returns, 5) * 100

    regime_hmm = hmm_results.get(ticker, {}).get("current_regime", "N/A")
    regime_gmm = gmm_results.get(ticker, {}).get("current_regime", "N/A")
    sentiment  = ticker_finbert.get(ticker, {})
    risk       = risk_metrics.get(ticker, {})
    mc         = mc_results.get(ticker, {})

    agent_rec  = agent_results.get(ticker, {})
    rec        = agent_rec.get("recommendation", "N/A")
    conf       = agent_rec.get("confidence",     0) * 100

    context = f"""
TICKER: {ticker}

MARKET DATA:
- Current Price    : ${last['Close']:.2f}
- Annual Return    : {ann_ret:.2f}%
- Annual Volatility: {ann_vol:.2f}%
- Sharpe Ratio     : {sharpe:.3f}
- Max Drawdown     : {max_dd:.2f}%
- RSI (14)         : {last['rsi']:.2f}
- MACD             : {last['macd']:.4f}
- VaR 95%          : {var_95:.2f}%

MARKET REGIME:
- HMM Regime       : {regime_hmm}
- GMM Regime       : {regime_gmm}

SENTIMENT:
- FinBERT Compound : {sentiment.get('compound', 0):.3f}
- Sentiment Label  : {sentiment.get('label', 'N/A')}

MONTE CARLO (252 days):
- Expected Return  : {mc.get('exp_return', 0):.1f}%
- P5 Price         : ${mc.get('p5', 0):.2f}
- P50 Price        : ${mc.get('p50', 0):.2f}
- P95 Price        : ${mc.get('p95', 0):.2f}
- Prob of Gain     : {mc.get('prob_gain', 0):.1f}%

AGENT RECOMMENDATION:
- Signal           : {rec}
- Confidence       : {conf:.0f}%
"""

    if not question or question.strip() == "":
        return context

    retrieval_q = f"{ticker} {question}"
    rag_docs    = hybrid_retriever.retrieve(retrieval_q, k=5)
    rag_ctx     = format_docs(rag_docs)

    time.sleep(3)
    prompt = f"""
You are FinSight AI Pro. Answer this question about {ticker} using the market data below.
Be specific with numbers and metrics.

MARKET DATA CONTEXT:
{context}

RAG CONTEXT:
{rag_ctx[:600]}

USER QUESTION: {question}

Answer:"""

    answer = call_llm(
        "You are FinSight AI Pro, an expert quantitative financial analyst.",
        prompt,
        max_tokens=512
    )
    return f"{context}\n\n{'='*60}\nAI ANALYSIS:\n{'='*60}\n{answer}"


def get_portfolio_summary():
    lines = ["=" * 65, "FINSIGHT AI PRO — PORTFOLIO SUMMARY", "=" * 65, ""]

    for ticker in CONFIG["market_tickers"]:
        df      = market_data[ticker]
        returns = df["returns"].dropna()
        ann_ret = returns.mean() * 252 * 100
        sharpe  = ann_ret / (returns.std() * np.sqrt(252) * 100 + 1e-9)
        regime  = hmm_results[ticker]["current_regime"]
        sent    = ticker_finbert.get(ticker, {}).get("label", "neutral")
        rec     = agent_results.get(ticker, {}).get("recommendation", "N/A")

        lines.append(
            f"{ticker:<6} | Ret={ann_ret:>7.2f}% | "
            f"Sharpe={sharpe:>6.3f} | "
            f"Regime={regime:<10} | "
            f"Sent={sent:<10} | "
            f"Signal={rec}"
        )

    lines.append("")
    lines.append("=" * 65)
    return "\n".join(lines)


portfolio_summary = get_portfolio_summary()
print()
print(portfolio_summary)

with gr.Blocks(title="FinSight AI Pro", theme=gr.themes.Soft()) as demo:
    gr.Markdown("# 🏦 FinSight AI Pro — Agentic RAG Financial Intelligence")
    gr.Markdown(
        "**10-Agent AI system** — BiLSTM + Ensemble + HMM Regime + "
        "FinBERT + Monte Carlo + RAG"
    )

    with gr.Tab("📊 Ticker Analysis"):
        with gr.Row():
            ticker_input   = gr.Dropdown(
                choices=CONFIG["market_tickers"],
                value="AAPL",
                label="Select Ticker"
            )
            question_input = gr.Textbox(
                label="Ask a question (optional)",
                placeholder="What is the current risk profile of this stock?",
                lines=2,
            )
        analyze_btn = gr.Button("🔍 Analyze", variant="primary")
        output_box  = gr.Textbox(label="Analysis", lines=30)
        analyze_btn.click(
            fn=analyze_ticker_gradio,
            inputs=[ticker_input, question_input],
            outputs=output_box,
        )

    with gr.Tab("📈 Portfolio Summary"):
        summary_box = gr.Textbox(
            value=portfolio_summary,
            label="Portfolio Overview",
            lines=20,
            interactive=False,
        )

    with gr.Tab("🤖 RAG Q&A"):
        rag_question = gr.Textbox(
            label="Financial Question",
            placeholder="Which stock has the best risk-adjusted return?",
            lines=2,
        )
        rag_btn    = gr.Button("💬 Ask AI", variant="primary")
        rag_output = gr.Textbox(label="AI Answer", lines=15)

        def rag_qa(question):
            if not question.strip():
                return "Please enter a question."
            docs    = hybrid_retriever.retrieve(question, k=5)
            context = format_docs(docs)
            time.sleep(3)
            answer  = call_llm(
                "You are FinSight AI Pro, an expert financial analyst. "
                "Use only the provided context.",
                f"Context:\n{context[:800]}\n\nQuestion: {question}",
                max_tokens=512,
            )
            return answer

        rag_btn.click(fn=rag_qa, inputs=rag_question, outputs=rag_output)

    with gr.Tab("ℹ️ System Info"):
        info = f"""
## FinSight AI Pro — System Information

**Architecture:**
- 10 Specialized AI Agents (LangGraph)
- BiLSTM + Attention (best DL model)
- XGBoost + LightGBM + CatBoost (gradient boosting)
- Weighted Ensemble + Ridge Stacking meta-learner
- HMM + GMM Market Regime Detection (3 regimes)
- FinBERT-style Custom Sentiment (BiGRU + Attention)
- Hybrid RAG (BM25 + FAISS, alpha=0.5)
- Monte Carlo (1000 simulations x 252 days)

**Data:**
- Tickers: {', '.join(CONFIG['market_tickers'])}
- Period : {CONFIG['start_date']} → {CONFIG['end_date']}
- Features: ~70 per ticker (OHLCV + Technical + Macro)

**Infrastructure:**
- Platform : Kaggle 2x Tesla T4 GPU
- Device   : {CONFIG['device']} ({CONFIG['n_gpus']} GPUs)
- LLM      : {CONFIG['llm_model']} via Groq
- Embedder : all-MiniLM-L6-v2 (custom)
"""
        gr.Markdown(info)

print("\nLaunching Gradio interface...")
demo.launch(share=True, quiet=True)

print()
print("=" * 65)
print("Cell 13 Summary — Final Dashboard + Gradio")
print("=" * 65)
print(f"Dashboard panels    : 9 (Price, Sharpe, VaR, MAPE, Corr,")
print(f"                         MC, Regime, Sentiment, Signals)")
print(f"Gradio tabs         : 4 (Analysis, Portfolio, RAG Q&A, Info)")
print(f"Tickers covered     : {len(CONFIG['market_tickers'])}")
print(f"Final report        : {CONFIG['results_dir'] / 'final_report.json'}")
print()
print("FinSight AI Pro — Project Complete!")
print()
print("=" * 65)
print("PIPELINE SUMMARY")
print("=" * 65)
print("Cell  1 : Installation")
print("Cell  2 : Imports + Config + API Keys + Custom Components")
print("Cell  3 : Market Data (10 tickers, 70+ features)")
print("Cell  4 : Macro Data + News + Sentiment")
print("Cell  5 : Corpus + Vector Store + Knowledge Graph")
print("Cell  6 : Advanced RAG (Dense + Hybrid + Multi-Query)")
print("Cell  7 : 10-Agent System (LangGraph)")
print("Cell  8 : LSTM + BiLSTM with Attention")
print("Cell  9 : XGBoost + LightGBM + CatBoost (Optuna)")
print("Cell 10 : Ensemble + Monte Carlo + VaR + Backtest")
print("Cell 11 : HMM + GMM Regime Detection")
print("Cell 12 : FinBERT Custom Sentiment")
print("Cell 13 : Dashboard + Gradio Interface")
print()
print("All cells complete. FinSight AI Pro is ready.")
```
